```python
import re
import unicodedata
import pandas as pd
from collections import Counter

ZERO_WIDTH_RE = re.compile(r"[\u200b\u200c]")

# Longest first to avoid partial stripping
BENGALI_SUFFIXES = [
    "গুলোতেই", "গুলোরই", "গুলোরও", "গুলোটাই",
    "গুলোতে", "গুলোয়", "গুলাতে", "গুলায়",
    "গুলোর", "গুলিকে", "গুলিতে", "গুলাদের", "গুলার",
    "দেরকেই", "দেরটাই", "দেরকে", "দেরটা", "দেরই", "দেরও",
    "গুলো", "গুলি", "গুলা",
    "দের", "রা", "গণ",
]

def normalize_unicode(text: str) -> str:
    return unicodedata.normalize("NFC", text)

def remove_zero_width(text: str) -> str:
    return ZERO_WIDTH_RE.sub("", text)

def normalize_spaces(text: str) -> str:
    text = re.sub(r"\s+", " ", text)
    return text.strip()

def strip_bangla_plural(token: str, min_stem_len: int = 2) -> str:
    for suf in BENGALI_SUFFIXES:
        if token.endswith(suf) and len(token) - len(suf) >= min_stem_len:
            return token[:-len(suf)]
    return token

def clean_text(text: str) -> str:
    text = normalize_unicode(str(text))
    text = remove_zero_width(text)
    text = normalize_spaces(text)
    tokens = text.split()
    tokens = [strip_bangla_plural(t) for t in tokens]
    return " ".join(tokens)

# Apply
df_clean["review_clean"] = df_clean["review"].apply(clean_text)
df_mal["review_clean"]   = df_mal["review"].apply(clean_text)
df_trust["review_clean"] = df_trust["review"].apply(clean_text)

print("Cleaning complete!")

# Sanity check
orig  = df_clean["review"].astype(str)
clean = df_clean["review_clean"].astype(str)
orig_vocab  = len(Counter(" ".join(orig).split()))
clean_vocab = len(Counter(" ".join(clean).split()))
print(f"Original vocab : {orig_vocab}")
print(f"After stripping: {clean_vocab}")
print(f"Reduction      : {100*(1 - clean_vocab/orig_vocab):.1f}%")
```

```python
import os
import gc
import random
from typing import Dict, List, Optional, Tuple

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    get_cosine_schedule_with_warmup,
)
from sklearn.utils.class_weight import compute_class_weight

try:
    from peft import get_peft_model, LoraConfig, TaskType
    PEFT_AVAILABLE = True
except ImportError:
    PEFT_AVAILABLE = False
    print("[Warning] peft is not installed. Run: pip install peft")


os.environ["TOKENIZERS_PARALLELISM"] = "false"

MODEL_NAME = "sagorsarker/bangla-bert-base"
SEED       = 42
DEVICE     = torch.device("cuda" if torch.cuda.is_available() else "cpu")

ENC_MAX_LEN        = 48
BATCH_SIZE         = 16
ACCUMULATION_STEPS = 2
LOCAL_EPOCHS       = 10
BASE_LR            = 5e-5
ADAPTER_LR         = 3e-4
WEIGHT_DECAY       = 0.05
WARMUP_RATIO       = 0.1

USE_LORA                    = True
LORA_R                      = 32
LORA_ALPHA                  = 64
LORA_DROPOUT                = 0.2
FREEZE_BOTTOM_N_LORA_LAYERS = 4

USE_FP16                   = True
USE_GRADIENT_CHECKPOINTING = False  # ← CHANGED: LoRA at 2% params doesn't need this

FUSION_BOTTLENECK     = 64
TRAIN_FUSION_AUX_LOSS = True
FUSION_AUX_WEIGHT     = 0.1

LABEL_SMOOTHING = 0.1

# ── Prototype loss weights (applied to CLS always) ──────────
CLS_COMPACTNESS_WEIGHT  = 0.1  # pull same-class CLS toward their center
CLS_SEPARATION_WEIGHT   = 0.1 # push class-0 center away from class-1 center
# ────────────────────────────────────────────────────────────

PROTOTYPE_CLASSES: List[int] = [0, 1]
MOM_K = 5

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if DEVICE.type == "cuda":
    torch.cuda.manual_seed_all(SEED)


# ============================================================
# Lazy tokenizer singleton
# ============================================================

_tokenizer: Optional[AutoTokenizer] = None

def get_tokenizer() -> AutoTokenizer:
    global _tokenizer
    if _tokenizer is None:
        _tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
    return _tokenizer


# ============================================================
# Helper Functions
# ============================================================

def get_bert_module(model: nn.Module) -> nn.Module:
    if hasattr(model, "bert"):
        return model.bert
    if hasattr(model, "base_model"):
        inner = model.base_model
        if hasattr(inner, "model"):
            real = inner.model
            if hasattr(real, "bert"):
                return real.bert
        if hasattr(inner, "bert"):
            return inner.bert
    for attr in ("model", "transformer", "encoder"):
        child = getattr(model, attr, None)
        if child is not None and hasattr(child, "bert"):
            return child.bert
    raise AttributeError(
        f"Could not find BERT backbone in model of type {type(model).__name__}.")


def get_classifier_module(model: nn.Module) -> nn.Module:
    if hasattr(model, "classifier"):
        return model.classifier
    if hasattr(model, "base_model"):
        inner = model.base_model
        if hasattr(inner, "model") and hasattr(inner.model, "classifier"):
            return inner.model.classifier
        if hasattr(inner, "classifier"):
            return inner.classifier
    if hasattr(model, "modules_to_save"):
        ms = model.modules_to_save
        if "classifier" in ms:
            mod = ms["classifier"]
            if hasattr(mod, "original_module"):
                return mod.original_module
            return mod
    raise AttributeError(
        f"Could not find classifier head in model of type {type(model).__name__}.")


def count_trainable_parameters(model: nn.Module) -> Tuple[int, int, float]:
    total     = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    pct       = 100.0 * trainable / max(total, 1)
    return trainable, total, pct


# ============================================================
# Model Initialization
# ============================================================

def freeze_bottom_lora_layers(
        model: nn.Module,
        freeze_bottom_n: int = FREEZE_BOTTOM_N_LORA_LAYERS) -> nn.Module:
    if freeze_bottom_n <= 0:
        return model
    frozen_count = 0
    for name, p in model.named_parameters():
        if "lora_" not in name:
            continue
        for layer_idx in range(freeze_bottom_n):
            patterns = [
                f"encoder.layer.{layer_idx}.",
                f"bert.encoder.layer.{layer_idx}.",
            ]
            if any(pattern in name for pattern in patterns):
                p.requires_grad = False
                frozen_count += p.numel()
                break
    print(f"[Freeze] Frozen LoRA params in bottom "
          f"{freeze_bottom_n} layers: {frozen_count:,}")
    return model


def init_banglabert_classifier(
    num_labels: int,
    use_lora: bool = USE_LORA,
    lora_r: int = LORA_R,
    lora_alpha: int = LORA_ALPHA,
    lora_dropout: float = LORA_DROPOUT,
    freeze_bottom_n_lora_layers: int = FREEZE_BOTTOM_N_LORA_LAYERS,
    use_gradient_checkpointing: bool = USE_GRADIENT_CHECKPOINTING,
    device: torch.device = DEVICE,
) -> nn.Module:
    model = AutoModelForSequenceClassification.from_pretrained(
        MODEL_NAME,
        num_labels=num_labels,
    )
    bert_backbone = get_bert_module(model)
    assert len(bert_backbone.encoder.layer) == 12, (
        f"Security violation: Found "
        f"{len(bert_backbone.encoder.layer)} layers. "
        "FedSemPro requires the full 12-layer BanglaBERT architecture."
    )
    if use_gradient_checkpointing and hasattr(
            model, "gradient_checkpointing_enable"):
        model.gradient_checkpointing_enable()
        if hasattr(model, "config"):
            model.config.use_cache = False
    if use_lora:
        if not PEFT_AVAILABLE:
            raise ImportError("PEFT is not installed. Run: pip install peft")
        lora_config = LoraConfig(
            task_type       = TaskType.SEQ_CLS,
            r               = lora_r,
            lora_alpha      = lora_alpha,
            target_modules  = [
                "query",
                "value",
                "key",
                "attention.output.dense",
                "intermediate.dense",
                "output.dense",
            ],
            lora_dropout    = lora_dropout,
            bias            = "none",
            modules_to_save = ["classifier"],
        )
        model = get_peft_model(model, lora_config)
        model = freeze_bottom_lora_layers(
            model, freeze_bottom_n=freeze_bottom_n_lora_layers)
    classifier = get_classifier_module(model)
    for p in classifier.parameters():
        p.requires_grad = True
    trainable, total, pct = count_trainable_parameters(model)
    print(f"[Model] Trainable params: {trainable:,} / {total:,} ({pct:.4f}%)")
    if use_lora and hasattr(model, "print_trainable_parameters"):
        model.print_trainable_parameters()
    return model.to(device)


# ============================================================
# Adaptive Fusion Head + Encoder
# ============================================================

class AdaptiveFusionHead(nn.Module):
    def __init__(self, hidden_size: int, bottleneck: int):
        super().__init__()
        self.adaptive_head = nn.Sequential(
            nn.Linear(hidden_size * 4, bottleneck),
            nn.LayerNorm(bottleneck),
            nn.GELU(),
            nn.Dropout(0.2),
            nn.Linear(bottleneck, 2),
        )
        self.fusion_norm = nn.LayerNorm(hidden_size)
        nn.init.zeros_(self.adaptive_head[-1].weight)
        nn.init.zeros_(self.adaptive_head[-1].bias)

    def forward(self, cls_emb: torch.Tensor, mean_emb: torch.Tensor):
        features = torch.cat(
            [cls_emb, mean_emb, cls_emb - mean_emb,
             cls_emb * mean_emb], dim=-1
        )
        weights = F.softmax(self.adaptive_head(features), dim=-1)
        alpha   = weights[:, 0:1]
        beta    = weights[:, 1:2]
        emb     = alpha * cls_emb + beta * mean_emb
        emb     = self.fusion_norm(emb)
        emb     = F.normalize(emb, p=2, dim=1)
        return emb, alpha, beta


class AdaptiveFusionEncoder(nn.Module):
    def __init__(
            self,
            base_model: nn.Module,
            max_len: int = ENC_MAX_LEN,
            bottleneck: int = FUSION_BOTTLENECK):
        super().__init__()
        self.base_model  = base_model
        self.bert        = get_bert_module(base_model)
        self.max_len     = int(max_len)
        hidden_size      = getattr(
            self.base_model.config, "hidden_size", 768)
        self.fusion_head = AdaptiveFusionHead(
            hidden_size=hidden_size, bottleneck=bottleneck)

    def fuse_from_hidden(
            self,
            hidden: torch.Tensor,
            attention_mask: torch.Tensor):
        cls_emb  = hidden[:, 0, :]
        mask     = attention_mask.unsqueeze(-1).float()
        mean_emb = (hidden * mask).sum(dim=1) / \
                   mask.sum(dim=1).clamp(min=1.0)
        return self.fusion_head(cls_emb, mean_emb)

    def forward_embeddings(
        self,
        input_ids: torch.Tensor,
        attention_mask: torch.Tensor,
        token_type_ids: Optional[torch.Tensor] = None,
    ):
        fwd: Dict = {
            "input_ids":      input_ids,
            "attention_mask": attention_mask,
            "return_dict":    True,
        }
        if token_type_ids is not None:
            fwd["token_type_ids"] = token_type_ids
        hidden = self.bert(**fwd).last_hidden_state
        return self.fuse_from_hidden(hidden, attention_mask)

    @torch.no_grad()
    def encode(
            self,
            texts: List[str],
            device: torch.device = DEVICE) -> torch.Tensor:
        self.eval()
        self.to(device)
        tok   = get_tokenizer()
        batch = tok(
            texts, padding=True, truncation=True,
            max_length=self.max_len,
            return_tensors="pt").to(device)
        emb, _, _ = self.forward_embeddings(
            input_ids=batch["input_ids"],
            attention_mask=batch["attention_mask"],
            token_type_ids=batch.get("token_type_ids"),
        )
        return emb

    @torch.no_grad()
    def inspect_fusion_weights(
            self,
            texts: List[str],
            device: torch.device = DEVICE) -> Dict[str, torch.Tensor]:
        self.eval()
        self.to(device)
        tok   = get_tokenizer()
        batch = tok(
            texts, padding=True, truncation=True,
            max_length=self.max_len,
            return_tensors="pt").to(device)
        emb, alpha, beta = self.forward_embeddings(
            input_ids=batch["input_ids"],
            attention_mask=batch["attention_mask"],
            token_type_ids=batch.get("token_type_ids"),
        )
        return {
            "alpha_cls":  alpha.squeeze(-1).cpu(),
            "beta_mean":  beta.squeeze(-1).cpu(),
            "embedding":  emb.cpu(),
        }


def init_text_encoder_from_model(
        model: nn.Module) -> AdaptiveFusionEncoder:
    return AdaptiveFusionEncoder(model, max_len=ENC_MAX_LEN)


# ============================================================
# Optimizer and Scheduler
# ============================================================

def get_optimizer(
    model: nn.Module,
    fusion_encoder: Optional[AdaptiveFusionEncoder] = None,
    base_lr: float = BASE_LR,
    adapter_lr: float = ADAPTER_LR,
    weight_decay: float = WEIGHT_DECAY,
) -> torch.optim.AdamW:
    no_decay = ["bias", "LayerNorm.weight"]
    lora_decay,   lora_nodecay   = [], []
    clf_decay,    clf_nodecay    = [], []
    fusion_decay, fusion_nodecay = [], []
    other_decay,  other_nodecay  = [], []

    for name, p in model.named_parameters():
        if not p.requires_grad:
            continue
        is_nodecay = any(nd in name for nd in no_decay)
        is_lora    = "lora_" in name
        is_clf     = "classifier" in name or "modules_to_save" in name
        if is_lora:
            (lora_nodecay if is_nodecay else lora_decay).append(p)
        elif is_clf:
            (clf_nodecay  if is_nodecay else clf_decay).append(p)
        else:
            (other_nodecay if is_nodecay else other_decay).append(p)

    if fusion_encoder is not None:
        for name, p in fusion_encoder.fusion_head.named_parameters():
            if not p.requires_grad:
                continue
            is_nodecay = any(nd in name for nd in no_decay)
            (fusion_nodecay if is_nodecay else fusion_decay).append(p)

    param_groups = []
    if lora_decay:     param_groups.append({"params": lora_decay,     "lr": adapter_lr, "weight_decay": weight_decay})
    if lora_nodecay:   param_groups.append({"params": lora_nodecay,   "lr": adapter_lr, "weight_decay": 0.0})
    if clf_decay:      param_groups.append({"params": clf_decay,      "lr": adapter_lr, "weight_decay": weight_decay})
    if clf_nodecay:    param_groups.append({"params": clf_nodecay,    "lr": adapter_lr, "weight_decay": 0.0})
    if fusion_decay:   param_groups.append({"params": fusion_decay,   "lr": adapter_lr, "weight_decay": weight_decay})
    if fusion_nodecay: param_groups.append({"params": fusion_nodecay, "lr": adapter_lr, "weight_decay": 0.0})
    if other_decay:    param_groups.append({"params": other_decay,    "lr": base_lr,    "weight_decay": weight_decay})
    if other_nodecay:  param_groups.append({"params": other_nodecay,  "lr": base_lr,    "weight_decay": 0.0})

    if not param_groups:
        raise ValueError("No trainable parameters found.")
    return torch.optim.AdamW(param_groups)


def get_scheduler(
    optimizer: torch.optim.Optimizer,
    num_epochs: int,
    steps_per_epoch: int,
    warmup_ratio: float = WARMUP_RATIO,
):
    total_steps  = max(1, num_epochs * steps_per_epoch)
    warmup_steps = max(1, int(warmup_ratio * total_steps))
    return get_cosine_schedule_with_warmup(
        optimizer,
        num_warmup_steps   = warmup_steps,
        num_training_steps = total_steps,
    )


# ============================================================
# FL Communication Utilities
# ============================================================

def get_trainable_state_dict(
    model: nn.Module,
    fusion_encoder: Optional[AdaptiveFusionEncoder] = None,
) -> Dict[str, torch.Tensor]:
    selected = {}
    for name, param in model.named_parameters():
        if param.requires_grad:
            selected[name] = param.detach().cpu().clone()
    if fusion_encoder is not None:
        for key, value in fusion_encoder.fusion_head.state_dict().items():
            selected[f"fusion_head.{key}"] = value.detach().cpu().clone()
    return selected


def split_model_and_fusion_state(
    state_dict: Dict[str, torch.Tensor],
) -> Tuple[Dict[str, torch.Tensor], Dict[str, torch.Tensor]]:
    model_state  = {}
    fusion_state = {}
    for key, value in state_dict.items():
        if key.startswith("fusion_head."):
            fusion_state[key.replace("fusion_head.", "", 1)] = value
        else:
            model_state[key] = value
    return model_state, fusion_state


def apply_trainable_state_dict(
    model: nn.Module,
    state_dict: Dict[str, torch.Tensor],
    fusion_encoder: Optional[AdaptiveFusionEncoder] = None,
) -> nn.Module:
    model_state, fusion_state = split_model_and_fusion_state(state_dict)
    param_dict = dict(model.named_parameters())
    for key, value in model_state.items():
        if key in param_dict:
            if param_dict[key].shape != value.shape:
                raise RuntimeError(f"Shape mismatch for '{key}'.")
            param_dict[key].data.copy_(
                value.to(param_dict[key].device))
    if fusion_encoder is not None and fusion_state:
        current_fusion = fusion_encoder.fusion_head.state_dict()
        for key, value in fusion_state.items():
            if key in current_fusion:
                if current_fusion[key].shape != value.shape:
                    raise RuntimeError(
                        f"Fusion head shape mismatch for '{key}'.")
        current_fusion.update({
            k: v.to(next(fusion_encoder.parameters()).device)
            for k, v in fusion_state.items()
            if k in current_fusion
        })
        fusion_encoder.fusion_head.load_state_dict(
            current_fusion, strict=False)
    return model


def aggregate_state_dicts(
    client_states: List[Dict[str, torch.Tensor]],
    client_sizes: Optional[List[int]] = None,
) -> Dict[str, torch.Tensor]:
    if not client_states:
        raise ValueError("client_states is empty.")
    n = len(client_states)
    if client_sizes is None:
        client_sizes = [1] * n
    total = float(sum(client_sizes))
    if total == 0:
        raise ValueError("Sum of client_sizes is zero.")
    weights = [s / total for s in client_sizes]
    keys    = client_states[0].keys()
    aggregated: Dict[str, torch.Tensor] = {}
    for key in keys:
        agg = sum(
            weights[i] * client_states[i][key].float()
            for i in range(n))
        aggregated[key] = agg.to(client_states[0][key].dtype)
    return aggregated


# ============================================================
# Prototype Losses
# ============================================================

def supervised_prototype_compactness_loss(
    embeddings: torch.Tensor,
    labels: torch.Tensor,
    eps: float = 1e-12,
) -> torch.Tensor:
    """Pull same-class embeddings toward their class center (cosine)."""
    losses        = []
    unique_labels = labels.detach().unique()
    for c in unique_labels:
        mask = labels == c
        if mask.sum() < 2:
            continue
        class_emb = embeddings[mask]
        center    = class_emb.mean(dim=0, keepdim=True)
        center    = F.normalize(center, p=2, dim=1, eps=eps)
        sim       = (class_emb * center).sum(dim=1)
        losses.append((1.0 - sim).mean())
    if not losses:
        return embeddings.new_tensor(0.0)
    return torch.stack(losses).mean()


def supervised_contrastive_separation_loss(
    embeddings: torch.Tensor,
    labels: torch.Tensor,
    margin: float = 0.5,
    eps: float = 1e-12,
) -> torch.Tensor:
    """Push class-0 center away from class-1 center (cosine margin)."""
    embeddings    = F.normalize(embeddings, p=2, dim=1, eps=eps)
    unique_labels = labels.detach().unique()
    if len(unique_labels) < 2:
        return embeddings.new_tensor(0.0)
    centers = []
    for c in unique_labels:
        mask   = labels == c
        center = F.normalize(
            embeddings[mask].mean(dim=0, keepdim=True), p=2, dim=1)
        centers.append(center)
    c0, c1 = centers[0], centers[1]
    sim    = (c0 * c1).sum()
    return F.relu(sim - margin)


# ============================================================
# Training
# ============================================================

_ce_loss = nn.CrossEntropyLoss(label_smoothing=LABEL_SMOOTHING)


def train_one_fl_round(
    model: nn.Module,
    train_loader,
    val_loader=None,
    fusion_encoder: Optional[AdaptiveFusionEncoder] = None,
    device: torch.device = DEVICE,
    local_epochs: int = LOCAL_EPOCHS,
    base_lr: float = BASE_LR,
    adapter_lr: float = ADAPTER_LR,
    weight_decay: float = WEIGHT_DECAY,
    accumulation_steps: int = ACCUMULATION_STEPS,
    use_fp16: bool = USE_FP16,
    train_fusion_aux_loss: bool = TRAIN_FUSION_AUX_LOSS,
    fusion_aux_weight: float = FUSION_AUX_WEIGHT,
    cls_compactness_weight: float = CLS_COMPACTNESS_WEIGHT,
    cls_separation_weight: float = CLS_SEPARATION_WEIGHT,
    cid: int = -1,
) -> Tuple[Dict[str, torch.Tensor], float, Dict]:
    # ── Adaptive LR: large clients get reduced LR ───────────
    total_samples = len(train_loader.dataset)
    if total_samples > 1500:
        adapter_lr = adapter_lr * 0.7
        base_lr    = base_lr   * 0.7
    print(f"[Client {cid}] N={total_samples} "
          f"→ adapter_lr={adapter_lr:.2e} base_lr={base_lr:.2e}")
    # ────────────────────────────────────────────────────────

    # ── Class-balanced loss ──────────────────────────────────
    try:
        labels_np = train_loader.dataset.labels
        cw = compute_class_weight(
            class_weight = 'balanced',
            classes      = np.array([0, 1]),
            y            = labels_np,
        )
        criterion = nn.CrossEntropyLoss(
            weight          = torch.tensor(cw, dtype=torch.float32).to(device),
            label_smoothing = LABEL_SMOOTHING,
        )
        print(f"[Client {cid}] Class weights: "
              f"0={cw[0]:.3f} 1={cw[1]:.3f}")
    except AttributeError:
        criterion = _ce_loss
        print(f"[Client {cid}] Using default loss "
              f"(dataset has no .labels attribute)")
    # ────────────────────────────────────────────────────────

    model.to(device)
    model.train()
    if fusion_encoder is not None:
        fusion_encoder.to(device)
        fusion_encoder.train()

    optimizer       = get_optimizer(
        model=model, fusion_encoder=fusion_encoder,
        base_lr=base_lr, adapter_lr=adapter_lr,
        weight_decay=weight_decay)
    steps_per_epoch = max(1, int(
        np.ceil(len(train_loader) / accumulation_steps)))
    scheduler       = get_scheduler(
        optimizer=optimizer, num_epochs=local_epochs,
        steps_per_epoch=steps_per_epoch)

    # ── CHANGED: use torch.amp instead of torch.cuda.amp ────
    _use_amp = use_fp16 and device.type == "cuda"
    scaler = torch.cuda.amp.GradScaler(enabled=_use_amp)
    # ────────────────────────────────────────────────────────

    use_fusion_loss = (
        fusion_encoder is not None and
        train_fusion_aux_loss and
        fusion_aux_weight > 0
    )

    train_loss_curve = []
    val_loss_curve   = []

    for epoch in range(local_epochs):
        model.train()
        if fusion_encoder is not None:
            fusion_encoder.train()

        epoch_loss  = 0.0
        epoch_steps = 0
        optimizer.zero_grad(set_to_none=True)

        for step, batch in enumerate(train_loader):
            input_ids      = batch["input_ids"].to(
                device, non_blocking=True)
            attention_mask = batch["attention_mask"].to(
                device, non_blocking=True)
            labels         = batch["labels"].to(
                device, non_blocking=True)

            fwd: Dict = {
                "input_ids":            input_ids,
                "attention_mask":       attention_mask,
                "output_hidden_states": True,
                "return_dict":          True,
            }
            if "token_type_ids" in batch:
                fwd["token_type_ids"] = batch[
                    "token_type_ids"].to(device, non_blocking=True)

            # ── CHANGED: use torch.amp.autocast ─────────────
            with torch.cuda.amp.autocast(enabled=_use_amp):
                out      = model(**fwd)
                raw_loss = criterion(out.logits, labels)

                # ── CLS embeddings (always available) ────────────
                hidden  = out.hidden_states[-1]           # (B, T, H)
                cls_emb = F.normalize(
                    hidden[:, 0, :], p=2, dim=1)         # (B, H)

                # compactness on CLS, every step
                compact_loss = supervised_prototype_compactness_loss(
                    cls_emb, labels)

                # separation on CLS, every step
                sep_loss = supervised_contrastive_separation_loss(
                    cls_emb, labels)

                raw_loss = (
                    raw_loss
                    + cls_compactness_weight * compact_loss
                    + cls_separation_weight  * sep_loss
                )

                # ── Optional fusion aux loss (unchanged) ─────────
                if use_fusion_loss:
                    emb, _, _ = fusion_encoder.fuse_from_hidden(
                        hidden, attention_mask)
                    fusion_loss = supervised_prototype_compactness_loss(
                        emb, labels)
                    raw_loss = raw_loss + fusion_aux_weight * fusion_loss
                # ─────────────────────────────────────────────────

                loss = raw_loss / accumulation_steps

            scaler.scale(loss).backward()
            epoch_loss  += float(raw_loss.detach().cpu())
            epoch_steps += 1

            should_step = (
                ((step + 1) % accumulation_steps == 0) or
                ((step + 1) == len(train_loader))
            )
            if should_step:
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(
                    model.parameters(), max_norm=1.0)
                if fusion_encoder is not None:
                    trainable_fusion = [
                        p for p in fusion_encoder.parameters()
                        if p.requires_grad]
                    if trainable_fusion:
                        torch.nn.utils.clip_grad_norm_(
                            trainable_fusion, max_norm=1.0)
                scaler.step(optimizer)
                scaler.update()
                scheduler.step()
                optimizer.zero_grad(set_to_none=True)

        avg_train_loss = epoch_loss / max(epoch_steps, 1)
        train_loss_curve.append(round(avg_train_loss, 6))

        # Val — logging only, no decisions made
        avg_val_loss = avg_train_loss
        if val_loader is not None:
            model.eval()
            val_loss_total = 0.0
            val_steps      = 0
            with torch.no_grad():
                for vbatch in val_loader:
                    vids  = vbatch["input_ids"].to(
                        device, non_blocking=True)
                    vmask = vbatch["attention_mask"].to(
                        device, non_blocking=True)
                    vlbl  = vbatch["labels"].to(
                        device, non_blocking=True)
                    vfwd  = {
                        "input_ids":      vids,
                        "attention_mask": vmask,
                        "return_dict":    True,
                    }
                    if "token_type_ids" in vbatch:
                        vfwd["token_type_ids"] = vbatch[
                            "token_type_ids"].to(
                            device, non_blocking=True)
                    # ── CHANGED: use torch.amp.autocast ─────
                    with torch.cuda.amp.autocast(enabled=_use_amp):
                        vout     = model(**vfwd)
                        val_loss = criterion(vout.logits, vlbl)
                    val_loss_total += float(val_loss.detach().cpu())
                    val_steps += 1
            avg_val_loss = val_loss_total / max(val_steps, 1)
            model.train()
            if fusion_encoder is not None:
                fusion_encoder.train()
            val_loss_curve.append(round(avg_val_loss, 6))
            print(f"[Client {cid}] Epoch {epoch+1}/{local_epochs} "
                  f"- Train Loss: {avg_train_loss:.4f} "
                  f"| Val Loss: {avg_val_loss:.4f}")
        else:
            print(f"[Client {cid}] Epoch {epoch+1}/{local_epochs} "
                  f"- Train Loss: {avg_train_loss:.4f}")

    avg_loss = float(np.mean(train_loss_curve)) \
        if train_loss_curve else 0.0

    history = {
        "train_loss_curve": train_loss_curve,
        "val_loss_curve":   val_loss_curve,
        "best_val_loss":    round(min(val_loss_curve) if val_loss_curve else 0.0, 6),
        "stopped_epoch":    local_epochs,
        "early_stopped":    0,
    }

    return get_trainable_state_dict(
        model, fusion_encoder=fusion_encoder), avg_loss, history


# ============================================================
# Balanced Proto Loader
# ============================================================
def make_balanced_proto_loader(
    dataset,
    labels_np: np.ndarray,
    batch_size: int = BATCH_SIZE,
    seed: int = SEED,
) -> torch.utils.data.DataLoader:
    rng  = np.random.default_rng(seed)
    idx0 = np.where(labels_np == 0)[0]
    idx1 = np.where(labels_np == 1)[0]

    # Use all samples from minority class
    n = min(len(idx0), len(idx1))

    if n == 0:
        raise ValueError("One class has no samples.")

    idx0_sampled = rng.choice(idx0, size=n, replace=False)
    idx1_sampled = rng.choice(idx1, size=n, replace=False)
    balanced_idx = np.concatenate([idx0_sampled, idx1_sampled])

    print(f"[ProtoLoader] class0={len(idx0)} class1={len(idx1)} "
          f"→ using {n} per class ({n*2} total)")

    subset = torch.utils.data.Subset(dataset, balanced_idx)
    return torch.utils.data.DataLoader(
        subset,
        batch_size = batch_size,
        shuffle    = False,
        drop_last  = False,
    )

# ============================================================
# MoM Prototype Computation
# ============================================================

def mom_prototype(
    E_np: np.ndarray,
    K: int = MOM_K,
    seed: int = SEED,
    eps: float = 1e-12,
) -> np.ndarray:
    n, dim = E_np.shape
    K_eff  = max(1, min(K, n))
    if K_eff <= 1 or n < 2:
        proto = E_np.mean(axis=0).astype(np.float32) \
            if n > 0 else np.zeros(dim, dtype=np.float32)
        norm  = np.linalg.norm(proto)
        return proto / (norm + eps)
    rng = np.random.default_rng(seed)
    idx = np.arange(n)
    rng.shuffle(idx)
    E_np   = E_np[idx]
    chunks = np.array_split(E_np, K_eff)
    means  = [c.mean(axis=0) for c in chunks if len(c) > 0]
    if not means:
        return np.zeros(dim, dtype=np.float32)
    means_stack = np.stack(means, axis=0)
    proto = np.median(means_stack, axis=0).astype(np.float32)
    norm  = np.linalg.norm(proto)
    return proto / (norm + eps)


@torch.no_grad()
def compute_class_prototypes_from_loader(
    model: nn.Module,
    data_loader,
    device: torch.device = DEVICE,
    classes: List[int] = PROTOTYPE_CLASSES,
    K: int = MOM_K,
    seed: int = SEED,
    eps: float = 1e-12,
) -> Dict[int, Optional[np.ndarray]]:
    """
    Extracts normalised CLS embeddings from last hidden layer,
    then computes MoM prototype per class.
    Normalisation before MoM ensures cosine similarity in score_ref
    measures direction not magnitude.
    Use make_balanced_proto_loader to ensure balanced input.
    """
    model.eval()
    model.to(device)
    emb_dict: Dict[int, List[np.ndarray]] = {
        int(c): [] for c in classes}

    for batch in data_loader:
        input_ids      = batch["input_ids"].to(device, non_blocking=True)
        attention_mask = batch["attention_mask"].to(
            device, non_blocking=True)
        labels         = batch["labels"].to(device, non_blocking=True)

        fwd: Dict = {
            "input_ids":            input_ids,
            "attention_mask":       attention_mask,
            "output_hidden_states": True,
            "return_dict":          True,
        }
        if "token_type_ids" in batch:
            fwd["token_type_ids"] = batch["token_type_ids"].to(
                device, non_blocking=True)

        outputs = model(**fwd)

        # CLS from last hidden layer — normalised before MoM
        cls_emb = outputs.hidden_states[-1][:, 0, :]          # (B, H)
        cls_emb = F.normalize(cls_emb, p=2, dim=1)            # unit norm
        emb_np  = cls_emb.detach().cpu().numpy().astype(np.float32)
        y_np    = labels.detach().cpu().numpy()

        for i, c in enumerate(y_np):
            c_int = int(c)
            if c_int in emb_dict:
                emb_dict[c_int].append(emb_np[i])

    prototypes: Dict[int, Optional[np.ndarray]] = {}
    for c in classes:
        c_int = int(c)
        vecs  = emb_dict[c_int]
        if len(vecs) == 0:
            prototypes[c_int] = None
            continue
        E_np = np.stack(vecs, axis=0).astype(np.float32)
        # vectors already unit-norm — MoM measures direction only
        prototypes[c_int] = mom_prototype(
            E_np=E_np, K=K,
            seed=seed + c_int, eps=eps).astype(np.float32)
    return prototypes


# ============================================================
# Cleanup
# ============================================================

def cleanup_cuda():
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.ipc_collect()


print("✅ Model utilities loaded successfully")
print(f"  Device               : {DEVICE}")
print(f"  LORA_R               : {LORA_R}")
print(f"  LORA_ALPHA           : {LORA_ALPHA}")
print(f"  LORA_DROPOUT         : {LORA_DROPOUT}")
print(f"  FREEZE               : {FREEZE_BOTTOM_N_LORA_LAYERS}")
print(f"  BASE_LR              : {BASE_LR}")
print(f"  ADAPTER_LR           : {ADAPTER_LR}  (×0.7 for N>1500)")
print(f"  WEIGHT_DECAY         : {WEIGHT_DECAY}")
print(f"  FUSION_AUX_WEIGHT    : {FUSION_AUX_WEIGHT}")
print(f"  CLS_COMPACTNESS_WT   : {CLS_COMPACTNESS_WEIGHT}")
print(f"  CLS_SEPARATION_WT    : {CLS_SEPARATION_WEIGHT}")
print(f"  LABEL_SMOOTHING      : {LABEL_SMOOTHING}")
print(f"  SCHEDULER            : cosine with warmup")
print("  LoRA targets         : query, value, key, attention.output.dense,")
print("                         intermediate.dense, output.dense")
print("  Prototype            : normalised CLS (hidden_states[-1][:, 0, :])")
print("  Proto loader         : balanced fixed sample via make_balanced_proto_loader")
print("  CLS compact+sep loss : applied every step, every client, every epoch")
print("  Early stopping       : REMOVED — full LOCAL_EPOCHS every round")
print("  Poisson epochs       : REMOVED — fixed LOCAL_EPOCHS")
print("  Fusion dropout       : 0.2")
print("  Adaptive LR          : N>1500 → ×0.7 | all others → default")
print("  Class weights        : balanced per client via compute_class_weight")
print("  FedProx              : REMOVED")
print("  Gradient checkpointing : DISABLED (LoRA 2% params — not needed)")
print("  AMP                  : torch.amp (updated from torch.cuda.amp)")
```

```python
from sklearn.model_selection import train_test_split
from collections import Counter
import numpy as np
import pandas as pd
from typing import Dict, List, Tuple, Optional


def split_non_iid_with_min_per_class(
    df: pd.DataFrame,
    client_ids: List[int],
    label_col: str,
    alpha: float,
    min_per_class: int,
    rng: np.random.RandomState,
) -> Dict[int, pd.DataFrame]:
    assert label_col in df.columns, f"df must have column '{label_col}'"

    # safer: ignore NaN + force int labels
    labels      = sorted(int(x) for x in df[label_col].dropna().unique())
    num_clients = len(client_ids)

    # sanity: nothing to split
    if num_clients == 0:
        return {}

    # hard min check per class
    for y in labels:
        n_y  = int((df[label_col] == y).sum())
        need = int(min_per_class * num_clients)
        if n_y < need:
            raise ValueError(
                f"[NonIID Split ERROR] Not enough samples for class {y}: {n_y} total, "
                f"need at least {need} (= {min_per_class} * {num_clients}) for clients {client_ids}.\n"
                f"Fix: reduce MIN_PER_CLASS or reduce #clients or oversample."
            )

    client_to_indices: Dict[int, List[int]] = {cid: [] for cid in client_ids}

    for y in labels:
        idxs = df.index[df[label_col] == y].to_numpy()
        rng.shuffle(idxs)

        base_needed = min_per_class * num_clients
        base        = idxs[:base_needed]
        remaining   = idxs[base_needed:]

        # hard minimum — every client gets min_per_class guaranteed
        start = 0
        for cid in client_ids:
            client_to_indices[cid].extend(base[start:start + min_per_class])
            start += min_per_class

        # dirichlet leftover
        if len(remaining) > 0:
            props  = rng.dirichlet(alpha * np.ones(num_clients))
            counts = np.floor(props * len(remaining)).astype(int)

            diff = len(remaining) - int(counts.sum())
            for i in range(diff):
                counts[i % num_clients] += 1

            start = 0
            for cid, c in zip(client_ids, counts):
                if c > 0:
                    client_to_indices[cid].extend(remaining[start:start + c])
                    start += c

    out: Dict[int, pd.DataFrame] = {}
    for cid in client_ids:
        idxs      = np.array(client_to_indices[cid])
        idxs      = rng.permutation(idxs)
        out[cid]  = df.loc[idxs].reset_index(drop=True)
    return out


# ============================================================
# ONE-TIME CACHE (stronger cache key)
# ============================================================
_NONIID_CACHE = {
    "key":               None,
    "client_datasets":   None,
    "df_global_test":    None,
    "honest_clients":    None,
    "malicious_clients": None,
}


def build_client_datasets(
    df_clean: pd.DataFrame,
    df_mal: pd.DataFrame,
    *,
    seed: int             = 42,
    num_clients: int      = 5,
    num_malicious: int    = 2,
    global_test_ratio: float = 0.30,   # ← 30% held out for server val+test
    alpha_honest: float   = 0.7,
    alpha_mal: float      = 0.7,
    min_per_class: int    = 400,
    label_col: str        = "y",
    verbose: bool         = False,
    use_cache: bool       = True,
) -> Tuple[Dict[int, pd.DataFrame], pd.DataFrame, List[int], List[int]]:
    """
    Splits df_clean and df_mal into per-client datasets.

    Data flow:
        df_clean (100%)
            70% → df_pool  → distributed to honest clients (Non-IID Dirichlet)
            30% → df_global_test → returned to server
                     server splits this 50/50:
                         15% → global_val_loader  (evaluate every round)
                         15% → global_test_loader (evaluate every round + final)

        df_mal (100%)
            100% → distributed to malicious clients (Non-IID Dirichlet)
    """
    assert 0 <= num_malicious < num_clients
    assert label_col in df_clean.columns, f"df_clean must have '{label_col}'"
    assert label_col in df_mal.columns,   f"df_mal must have '{label_col}'"

    # stronger fingerprint: same length but different shuffle/labels won't reuse cache wrongly
    finger_clean = int(pd.util.hash_pandas_object(df_clean[label_col], index=False).sum())
    finger_mal   = int(pd.util.hash_pandas_object(df_mal[label_col],   index=False).sum())

    cache_key = (
        seed, num_clients, num_malicious, global_test_ratio,
        alpha_honest, alpha_mal, min_per_class, label_col,
        int(len(df_clean)), int(len(df_mal)),
        finger_clean, finger_mal,
    )

    if use_cache and _NONIID_CACHE["key"] == cache_key:
        return (
            _NONIID_CACHE["client_datasets"],
            _NONIID_CACHE["df_global_test"],
            _NONIID_CACHE["honest_clients"],
            _NONIID_CACHE["malicious_clients"],
        )

    rng = np.random.RandomState(seed)

    honest_clients    = list(range(0, num_clients - num_malicious))
    malicious_clients = list(range(num_clients - num_malicious, num_clients))

    # ── Global hold-out (no leakage) ────────────────────────
    # 30% of df_clean reserved for server val + test
    # 70% distributed to honest clients
    df_clean_ = df_clean.reset_index(drop=True)
    df_pool, df_global_test = train_test_split(
        df_clean_,
        test_size    = global_test_ratio,   # 0.30
        random_state = seed,
        shuffle      = True,
        stratify     = df_clean_[label_col],
    )
    df_pool        = df_pool.reset_index(drop=True)
    df_global_test = df_global_test.reset_index(drop=True)

    if verbose:
        print("=" * 50)
        print("Building client datasets (NON-IID Dirichlet + min-per-class)")
        print(f"NUM_CLIENTS       : {num_clients}")
        print(f"NUM_MALICIOUS     : {num_malicious}")
        print(f"Honest clients    : {honest_clients}")
        print(f"Malicious clients : {malicious_clients}")
        print(f"GLOBAL_TEST_RATIO : {global_test_ratio}  "
              f"(server val+test = {global_test_ratio/2:.2f} each)")
        print(f"MIN_PER_CLASS     : {min_per_class}")
        print(f"ALPHA_HONEST      : {alpha_honest}")
        print(f"ALPHA_MAL         : {alpha_mal}")
        print("=" * 50)
        print(f"\nGlobal hold-out  : {df_global_test.shape}  "
              f"(→ {len(df_global_test)//2} val + {len(df_global_test)//2} test)")
        print(f"Client pool      : {df_pool.shape}")

    # ── Non-IID split for honest clients ────────────────────
    clean_client_datasets = split_non_iid_with_min_per_class(
        df           = df_pool,
        client_ids   = honest_clients,
        label_col    = label_col,
        alpha        = alpha_honest,
        min_per_class= min_per_class,
        rng          = rng,
    )

    # ── Non-IID split for malicious clients ─────────────────
    mal_client_datasets = split_non_iid_with_min_per_class(
        df           = df_mal.reset_index(drop=True),
        client_ids   = malicious_clients,
        label_col    = label_col,
        alpha        = alpha_mal,
        min_per_class= min_per_class,
        rng          = rng,
    )

    client_datasets: Dict[int, pd.DataFrame] = {}
    client_datasets.update(clean_client_datasets)
    client_datasets.update(mal_client_datasets)

    if verbose:
        print("\nClient sizes:", [len(client_datasets[cid]) for cid in range(num_clients)])
        for i in range(num_clients):
            cdf   = client_datasets[i]
            cnt   = Counter(cdf[label_col].tolist())
            total = len(cdf)
            role  = "HON" if i in honest_clients else "MAL"
            print(
                f"Client {i} [{role}]: N={total} -> "
                + ", ".join(f"{k}:{v}({v/total:.2f})"
                            for k, v in sorted(cnt.items()))
            )

    if use_cache:
        _NONIID_CACHE["key"]               = cache_key
        _NONIID_CACHE["client_datasets"]   = client_datasets
        _NONIID_CACHE["df_global_test"]    = df_global_test
        _NONIID_CACHE["honest_clients"]    = honest_clients
        _NONIID_CACHE["malicious_clients"] = malicious_clients

    return client_datasets, df_global_test, honest_clients, malicious_clients
```

```python
from transformers import AutoTokenizer
from typing import Dict, List

BASE_MODEL_NAME = "sagorsarker/bangla-bert-base" 
MODEL_NAME = BASE_MODEL_NAME


def detect_vocab_holes(vocab_dict: Dict[str, int], tokenizer_len: int) -> List[int]:
    """
    Check whether tokenizer vocabulary indices contain missing positions.
    This is only a warning check. We do not rebuild tokenizer,
    because rebuilding can change original BanglaBERT tokenization behavior.
    """
    if not vocab_dict:
        return []

    # ✅ Fix 1: Use a set for O(1) lookup speed, avoiding O(N^2) bottlenecks
    idx_set = set(vocab_dict.values())
    
    # ✅ Fix 2: Safeguard the upper limit using len(tokenizer) to account for any added tokens
    max_boundary = max(max(idx_set) if idx_set else 0, tokenizer_len)

    holes = [i for i in range(max_boundary) if i not in idx_set]
    return holes


print("\n--- Loading BanglaBERT tokenizer ---")

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    use_fast=True
)

try:
    vocab = tokenizer.get_vocab()
    # Pass tokenizer length to ensure edge-case tokens aren't missed
    holes = detect_vocab_holes(vocab, len(tokenizer))

    if holes:
        print(f"Warning: tokenizer vocab has {len(holes)} holes.")
        print("Original BanglaBERT tokenizer is still used to preserve pretrained tokenization.")
    else:
        print("No tokenizer vocab holes detected.")

except Exception as e:
    print(f"Tokenizer vocabulary check skipped due to error: {e}")

print("Tokenizer loaded successfully.")
print("Tokenizer vocab size:", len(tokenizer))
```

```python
import gc
import json
import logging
from typing import Dict, Tuple, List, Optional

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
import flwr as fl

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

DEVICE      = torch.device("cuda" if torch.cuda.is_available() else "cpu")
BATCH_SIZE  = 16
ENC_MAX_LEN = 48
NUM_LABELS  = 2


# ============================================================
# Dataset Class
# ============================================================

class BanglaSentimentDataset(Dataset):
    def __init__(self, df: pd.DataFrame, text_col: str,
                 label_col: str, max_len: int = 48):
        self.texts   = df[text_col].astype(str).tolist()
        self.labels  = df[label_col].astype(int).tolist()
        self.tok     = get_tokenizer()
        self.max_len = max_len

    def __len__(self) -> int:
        return len(self.texts)

    def __getitem__(self, idx: int) -> Dict[str, torch.Tensor]:
        enc  = self.tok(
            self.texts[idx],
            truncation=True,
            padding="max_length",
            max_length=self.max_len,
            return_tensors="pt",
        )
        item = {k: v.squeeze(0) for k, v in enc.items()}
        item["labels"] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item


# ============================================================
# FedSemClient — train only, server-side evaluation
# ============================================================

class FedSemClient(fl.client.NumPyClient):

    def __init__(
            self,
            cid: int,
            model: nn.Module,
            encoder: nn.Module,
            train_df: pd.DataFrame,
            val_df: Optional[pd.DataFrame],
            test_df: Optional[pd.DataFrame],
            num_labels: int = NUM_LABELS,
        ):
            self.cid        = int(cid)
            self.num_labels = num_labels
            self.model      = model.to(DEVICE)
            self.encoder    = encoder.to(DEVICE)
    
            with torch.no_grad():
                init_state      = get_trainable_state_dict(
                    self.model, fusion_encoder=self.encoder)
                self.param_keys = list(init_state.keys())
            del init_state
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
    
            ds_train = BanglaSentimentDataset(
                train_df, "review_clean", "y", ENC_MAX_LEN)
    
            # ── Balanced train loader ────────────────────────────
            labels_np      = np.array(ds_train.labels)
            class_counts   = np.bincount(labels_np)
            class_weights  = 1.0 / class_counts
            sample_weights = class_weights[labels_np]
            sampler        = WeightedRandomSampler(
                weights     = torch.FloatTensor(sample_weights),
                num_samples = len(sample_weights),
                replacement = True,
            )
            self.train_loader = DataLoader(
                ds_train,
                batch_size = BATCH_SIZE,
                sampler    = sampler,
                drop_last  = False,
            )
    
            # ── Val loader — clean, deterministic, no sampler ───
            # Logging only — no decisions made on val loss
            if val_df is not None:
                ds_val          = BanglaSentimentDataset(
                    val_df, "review_clean", "y", ENC_MAX_LEN)
                self.val_loader = DataLoader(
                    ds_val,
                    batch_size = BATCH_SIZE,
                    shuffle    = False,
                    drop_last  = False,
                )
                self.num_val_samples = len(ds_val)
            else:
                self.val_loader      = None
                self.num_val_samples = 0
    
            # ── Test loader — real distribution, no shuffle ─────
            if test_df is not None:
                ds_test          = BanglaSentimentDataset(
                    test_df, "review_clean", "y", ENC_MAX_LEN)
                self.test_loader = DataLoader(
                    ds_test,
                    batch_size = BATCH_SIZE,
                    shuffle    = False,
                    drop_last  = False,
                )
                self.num_test_samples = len(ds_test)
            else:
                self.test_loader      = None
                self.num_test_samples = 0
    
            # ── Proto loader — balanced fixed sample ─────────────
            # Balanced 50/50 subset so proto centroid is not pulled
            # toward majority class. Comparable to mu_ref (249/249).
            self.proto_loader = make_balanced_proto_loader(
                dataset    = ds_train,
                labels_np  = labels_np,
                batch_size = BATCH_SIZE,
            )
    
            self.num_train_samples = len(ds_train)

    # ----------------------------------------------------------
    def get_parameters(self, config):
        state_dict = get_trainable_state_dict(
            self.model, fusion_encoder=self.encoder)
        return [state_dict[k].cpu().numpy() for k in self.param_keys]

    # ----------------------------------------------------------
    def set_parameters(self, parameters):
        state_dict = {
            k: torch.tensor(v, device=DEVICE)
            for k, v in zip(self.param_keys, parameters)
        }
        apply_trainable_state_dict(
            self.model, state_dict, fusion_encoder=self.encoder)
        first_val = list(state_dict.values())[0]
        print(f"[Client {self.cid}] set_params "
              f"first_mean={first_val.float().mean():.6f}")

    # ----------------------------------------------------------
    def fit(self, parameters, config):
        # Capture incoming weights BEFORE applying
        init_params_np = [v.copy() for v in parameters]

        self.set_parameters(parameters)

        epochs = int(config.get("epochs", LOCAL_EPOCHS))

        final_state, avg_loss, history = train_one_fl_round(
            model                  = self.model,
            train_loader           = self.train_loader,
            val_loader             = self.val_loader,
            fusion_encoder         = self.encoder,
            device                 = DEVICE,
            local_epochs           = epochs,
            base_lr                = BASE_LR,
            adapter_lr             = ADAPTER_LR,
            weight_decay           = WEIGHT_DECAY,
            accumulation_steps     = ACCUMULATION_STEPS,
            use_fp16               = USE_FP16,
            cls_compactness_weight = CLS_COMPACTNESS_WEIGHT,  # ← new
            cls_separation_weight  = CLS_SEPARATION_WEIGHT,   # ← new
            cid                    = self.cid,
        )

        apply_trainable_state_dict(
            self.model, final_state, fusion_encoder=self.encoder)
        final_params_np = self.get_parameters(config={})

        # Compute update L2
        update_l2 = float(sum(
            float(np.sum(
                (fp.astype(np.float32) - ip.astype(np.float32)) ** 2))
            for fp, ip in zip(final_params_np, init_params_np)
        ) ** 0.5)
        print(f"[Client {self.cid}] update_l2 = {update_l2:.4f}")

        # Compute train accuracy
        self.model.eval()
        train_correct = 0
        train_total   = 0
        with torch.no_grad():
            for batch in self.train_loader:
                input_ids      = batch["input_ids"].to(DEVICE)
                attention_mask = batch["attention_mask"].to(DEVICE)
                labels         = batch["labels"].to(DEVICE)
                fwd = {
                    "input_ids":      input_ids,
                    "attention_mask": attention_mask,
                    "return_dict":    True,
                }
                if "token_type_ids" in batch:
                    fwd["token_type_ids"] = batch["token_type_ids"].to(DEVICE)
                with torch.cuda.amp.autocast(enabled=DEVICE.type == "cuda"):
                    out = self.model(**fwd)
                preds          = torch.argmax(out.logits, dim=1)
                train_correct += int((preds == labels).sum().item())
                train_total   += int(labels.size(0))
        self.model.train()
        train_acc = float(train_correct / max(train_total, 1))

        val_loss, val_acc   = self._evaluate_local(use_test=False)
        test_loss, test_acc = self._evaluate_local(use_test=True)

        # Proto computed server-side — send empty
        proto_metrics = {
            "has_protos":        0,
            "proto_sep_score":   0.0,
            "proto_missing_0":   1,
            "proto_missing_1":   1,
            "proto_0_vec_json":  "",
            "proto_1_vec_json":  "",
            "proto_n0":          0,
            "proto_n1":          0,
            "proto_0_norm":      0.0,
            "proto_1_norm":      0.0,
            "proto_sep_cos":     0.0,
        }

        metrics = {
            "cid":                   int(self.cid),
            "n_train":               int(self.num_train_samples),
            "n_val":                 int(self.num_val_samples),
            "n_test":                int(self.num_test_samples),
            "train_loss":            float(avg_loss),
            "train_acc":             float(train_acc),
            "val_loss":              float(val_loss),
            "val_acc":               float(val_acc),
            "test_loss":             float(test_loss),
            "test_acc":              float(test_acc),
            "update_l2":             float(update_l2),
            "avg_local_loss":        float(avg_loss),
            "train_loss_curve_json": json.dumps(
                history["train_loss_curve"]),
            "val_loss_curve_json":   json.dumps(
                history["val_loss_curve"]),
            "best_val_loss":         float(history["best_val_loss"]),
            "stopped_epoch":         int(history["stopped_epoch"]),
            "early_stopped":         int(history["early_stopped"]),
        }
        metrics.update(proto_metrics)

        gc.collect()
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

        return self.get_parameters(config={}), self.num_train_samples, metrics

    # ----------------------------------------------------------
    def evaluate(self, parameters, config):
        # Client-side evaluation disabled.
        # Global model is evaluated centrally at the server
        # via evaluate_fn in FedSemDefenceStrategy.
        # This method is kept for Flower compatibility only.
        return 0.0, self.num_test_samples, {"accuracy": 0.0}

    # ----------------------------------------------------------
    def _evaluate_local(self, use_test: bool = False) -> Tuple[float, float]:
            self.model.eval()
            loader        = self.test_loader if use_test else self.val_loader
            total_samples = (self.num_test_samples if use_test
                             else self.num_val_samples)
    
            # guard — loader is None when no val/test data provided
            if loader is None or total_samples == 0:
                return 0.0, 0.0
    
            running_loss  = 0.0
            correct_preds = 0
            _use_amp      = DEVICE.type == "cuda"
            criterion     = nn.CrossEntropyLoss()
    
            with torch.no_grad():
                for batch in loader:
                    input_ids      = batch["input_ids"].to(DEVICE)
                    attention_mask = batch["attention_mask"].to(DEVICE)
                    labels         = batch["labels"].to(DEVICE)
                    fwd = {
                        "input_ids":      input_ids,
                        "attention_mask": attention_mask,
                        "return_dict":    True,
                    }
                    if "token_type_ids" in batch:
                        fwd["token_type_ids"] = batch["token_type_ids"].to(DEVICE)
                    with torch.cuda.amp.autocast(enabled=_use_amp):
                        out  = self.model(**fwd)
                        loss = criterion(out.logits, labels)
                    running_loss  += (float(loss.detach().cpu())
                                      * input_ids.size(0))
                    preds          = torch.argmax(out.logits, dim=1)
                    correct_preds += int((preds == labels).sum().item())
    
            return (running_loss / max(total_samples, 1),
                correct_preds / max(total_samples, 1))

    # ----------------------------------------------------------
    def _calculate_semantic_prototypes(self) -> Dict:
        self.model.eval()
        embeddings_list = []
        labels_list     = []

        with torch.no_grad():
            for batch in self.proto_loader:
                input_ids      = batch["input_ids"].to(DEVICE)
                attention_mask = batch["attention_mask"].to(DEVICE)
                labels         = batch["labels"].to(DEVICE)

                fwd = {
                    "input_ids":            input_ids,
                    "attention_mask":       attention_mask,
                    "output_hidden_states": True,
                    "return_dict":          True,
                }
                if "token_type_ids" in batch:
                    fwd["token_type_ids"] = batch["token_type_ids"].to(DEVICE)

                outputs = self.model(**fwd)

                # CLS from last hidden layer — normalised for cosine consistency
                cls_emb = outputs.hidden_states[-1][:, 0, :]
                cls_emb = F.normalize(cls_emb, p=2, dim=1)

                embeddings_list.append(cls_emb.cpu().numpy())
                labels_list.append(labels.cpu().numpy())

        if not embeddings_list:
            return {"has_protos": 0, "proto_sep_score": 0.0,
                    "proto_missing_0": 1, "proto_missing_1": 1}

        all_embeddings = np.concatenate(
            embeddings_list, axis=0).astype(np.float32)
        all_labels     = np.concatenate(labels_list, axis=0)

        class_0_emb = all_embeddings[all_labels == 0]
        class_1_emb = all_embeddings[all_labels == 1]

        missing_0 = class_0_emb.shape[0] == 0
        missing_1 = class_1_emb.shape[0] == 0

        if missing_0 or missing_1:
            return {
                "has_protos":      0,
                "proto_sep_score": 0.0,
                "proto_missing_0": int(missing_0),
                "proto_missing_1": int(missing_1),
                "proto_n0":        int(class_0_emb.shape[0]),
                "proto_n1":        int(class_1_emb.shape[0]),
            }

        proto_0 = mom_prototype(class_0_emb, K=MOM_K, seed=SEED)
        proto_1 = mom_prototype(class_1_emb, K=MOM_K, seed=SEED + 1)

        cos_sim = float(
            np.dot(proto_0, proto_1) /
            (np.linalg.norm(proto_0) * np.linalg.norm(proto_1) + 1e-12))
        proto_sep_score = float(1.0 - np.clip(cos_sim, -1.0, 1.0))

        return {
            "has_protos":        1,
            "proto_sep_score":   proto_sep_score,
            "proto_missing_0":   0,
            "proto_missing_1":   0,
            "proto_n0":          int(class_0_emb.shape[0]),
            "proto_n1":          int(class_1_emb.shape[0]),
            "proto_0_norm":      float(np.linalg.norm(proto_0)),
            "proto_1_norm":      float(np.linalg.norm(proto_1)),
            "proto_sep_cos":     float(cos_sim),
            "proto_0_vec_json":  json.dumps(proto_0.tolist()),
            "proto_1_vec_json":  json.dumps(proto_1.tolist()),
        }


print("✅ BanglaSentimentDataset and FedSemClient defined successfully")
```

```python
from typing import List, Tuple, Dict, Any, Optional
from collections import defaultdict
import json

import flwr as fl
import numpy as np
import torch
import torch.nn as nn
from flwr.server.strategy import FedAvg
from flwr.common import parameters_to_ndarrays


# ============================================================
# Global Histories
# ============================================================

FEDSEM_HIST: Dict[str, Dict] = {
    "server": {
        "round": [],

        "train_loss_w": [],
        "train_acc_w": [],

        "val_loss_w": [],
        "val_acc_w": [],
        "val_precision_w": [],
        "val_recall_w": [],
        "val_f1_w": [],

        "test_loss_w": [],
        "test_acc_w": [],
        "test_precision_w": [],
        "test_recall_w": [],
        "test_f1_w": [],

        "avg_local_loss_w": [],
        "early_stopped_count": [],
        "active_clients": [],
        "banned_clients": [],
        "failures": [],

        # ── Global centralized evaluation (server-side) ──────
        "global_val_loss":  [],
        "global_val_acc":   [],
        "global_test_loss": [],
        "global_test_acc":  [],
        # ────────────────────────────────────────────────────
    },
    "clients": {},
}

FEDSEM_GLOBAL_FINAL = {"params": None}


# ============================================================
# Utility Helpers
# ============================================================

def _safe_float(x: Any, default: float = 0.0) -> float:
    try:
        if x is None:
            return default
        v = float(x)
        if not np.isfinite(v):
            return default
        return v
    except Exception:
        return default


def _safe_int(x: Any, default: int = 0) -> int:
    try:
        if x is None:
            return default
        return int(x)
    except Exception:
        return default


def _safe_json_or_none(x: Any) -> Optional[str]:
    if x is None:
        return None
    if isinstance(x, str):
        return x
    try:
        return json.dumps(x)
    except Exception:
        return None


def _init_client_history(cid: str) -> None:
    if cid in FEDSEM_HIST["clients"]:
        return

    FEDSEM_HIST["clients"][cid] = {k: [] for k in [
        "round",

        # training/evaluation
        "train_loss",
        "train_acc",
        "val_loss",
        "val_acc",
        "val_precision",
        "val_recall",
        "val_f1",
        "val_cm_json",
        "test_loss",
        "test_acc",
        "test_precision",
        "test_recall",
        "test_f1",
        "test_cm_json",
        "avg_local_loss",

        # client sizes/status
        "n_train",
        "n_val",
        "n_test",
        "share_weight",
        "tau",
        "banned_client",

        # early stopping + graph drawing
        "loss_history_json",
        "train_loss_curve_json",
        "val_loss_curve_json",
        "early_stopped",
        "stopped_epoch",
        "best_val_loss",

        # update and prototypes
        "update_l2",
        "has_protos",
        "sep_score_client",
        "proto_missing_0",
        "proto_missing_1",
        "proto_0_norm",
        "proto_1_norm",
        "proto_sep_cos",
        "proto_sep_angle",
        "proto_sep_weight",
        "proto_sep_score",
        "proto_n0",
        "proto_n1",

        # optional vector logging
        "proto_0_vec_json",
        "proto_1_vec_json",
    ]}


# ============================================================
# get_evaluate_fn — server-side centralized evaluation
# ============================================================

def get_evaluate_fn(model, encoder, val_loader, test_loader, device, param_keys):
    """
    Returns an evaluate_fn for Flower strategy.
    Called automatically after every round by Flower.
    Evaluates global model on clean centralized val and test sets.

    Following standard Byzantine-robust FL evaluation protocol:
    McMahan et al. (2017), FLTrust Cao et al. (2022), DDaBA (2022).
    Client-side evaluation is disabled — malicious clients have
    poisoned local data making client metrics unreliable.
    """
    def evaluate(server_round: int, parameters, config):

        # Load aggregated parameters into global model
        state_dict = {
            k: torch.tensor(v, device=device)
            for k, v in zip(param_keys, parameters)
        }
        apply_trainable_state_dict(model, state_dict, fusion_encoder=encoder)
        model.eval()

        criterion = nn.CrossEntropyLoss()

        def run_eval(loader):
            correct  = 0
            total    = 0
            loss_sum = 0.0
            _use_amp = device.type == "cuda"
            with torch.no_grad():
                for batch in loader:
                    ids  = batch["input_ids"].to(device)
                    mask = batch["attention_mask"].to(device)
                    lbls = batch["labels"].to(device)
                    fwd  = {
                        "input_ids":      ids,
                        "attention_mask": mask,
                        "return_dict":    True,
                    }
                    if "token_type_ids" in batch:
                        fwd["token_type_ids"] = batch["token_type_ids"].to(device)
                    with torch.cuda.amp.autocast(enabled=_use_amp):
                        out  = model(**fwd)
                        loss = criterion(out.logits, lbls)
                    loss_sum += float(loss.detach().cpu()) * ids.size(0)
                    preds     = torch.argmax(out.logits, dim=1)
                    correct  += int((preds == lbls).sum().item())
                    total    += int(lbls.size(0))
            return (
                loss_sum / max(total, 1),
                correct  / max(total, 1),
            )

        val_loss,  val_acc  = run_eval(val_loader)
        test_loss, test_acc = run_eval(test_loader)

        print(
            f"[Global Round {server_round}] "
            f"Val  Loss={val_loss:.4f} Acc={val_acc:.4f} | "
            f"Test Loss={test_loss:.4f} Acc={test_acc:.4f}"
        )

        # Store in FEDSEM_HIST for plotting
        FEDSEM_HIST["server"]["global_val_loss"].append(round(val_loss,  4))
        FEDSEM_HIST["server"]["global_val_acc"].append(round(val_acc,   4))
        FEDSEM_HIST["server"]["global_test_loss"].append(round(test_loss, 4))
        FEDSEM_HIST["server"]["global_test_acc"].append(round(test_acc,  4))

        # Save to jsonl for easy plotting
        with open("global_eval_history.jsonl", "a") as f:
            f.write(json.dumps({
                "round":     server_round,
                "val_loss":  round(val_loss,  4),
                "val_acc":   round(val_acc,   4),
                "test_loss": round(test_loss, 4),
                "test_acc":  round(test_acc,  4),
            }) + "\n")

        return val_loss, {
            "val_accuracy":  val_acc,
            "val_loss":      val_loss,
            "test_accuracy": test_acc,
            "test_loss":     test_loss,
        }

    return evaluate


# ============================================================
# Recording FedAvg Strategy
# ============================================================

class RecordingFedSem(FedAvg):
    """
    Logger-only FedAvg.

    Performs standard FedAvg aggregation.
    Records per-client and server-level metrics every round.
    Global model evaluated centrally via evaluate_fn — not on clients.

    Usage:
        strategy = RecordingFedSem(
            ...
            evaluate_fn = get_evaluate_fn(
                model       = global_model,
                encoder     = global_encoder,
                val_loader  = global_val_loader,
                test_loader = global_test_loader,
                device      = DEVICE,
                param_keys  = param_keys,
            ),
        )
    """

    def __init__(
        self,
        *args,
        title: str = "FedSem Logger (FedAvg baseline, LoRA/adaptive metrics)",
        print_client_table: bool = True,
        **kwargs,
    ):
        super().__init__(*args, **kwargs)
        self.title = str(title)
        self.print_client_table = bool(print_client_table)

    def on_round_begin(self, server_round: int) -> None:
        print(f"\n==================== FedSem Round {server_round} ====================")

    def configure_fit(
        self,
        server_round: int,
        parameters: fl.common.Parameters,
        client_manager: fl.server.client_manager.ClientManager,
    ):
        cfg = super().configure_fit(server_round, parameters, client_manager)

        new_cfg = []
        for client_proxy, fit_ins in cfg:
            c = dict(fit_ins.config) if fit_ins.config is not None else {}
            c["server_round"] = int(server_round)
            c["epochs"]       = LOCAL_EPOCHS   # fixed epochs — no poisson
            c.setdefault("banned_clients", [])
            c.setdefault("do_train_print", 0)
            new_cfg.append((client_proxy, fl.common.FitIns(fit_ins.parameters, c)))

        return new_cfg

    def configure_evaluate(
        self,
        server_round: int,
        parameters: fl.common.Parameters,
        client_manager: fl.server.client_manager.ClientManager,
    ):
        # Client-side evaluation disabled.
        # Global model evaluated centrally via evaluate_fn.
        return []

    def aggregate_fit(
        self,
        server_round: int,
        results: List[Tuple[fl.server.client_proxy.ClientProxy, fl.common.FitRes]],
        failures: List[Any],
    ):
        if not results:
            print(f"[Server][Round {server_round}] No fit results received.")
            return None, {}

        agg = super().aggregate_fit(server_round, results, failures)
        if agg is None:
            print(f"[Server][Round {server_round}] Super aggregation returned None.")
            return None, {}

        new_params, _metrics_from_super = agg

        def wavg(key: str, default: float = 0.0) -> float:
            numerator = 0.0
            denom     = 0.0
            for _, res in results:
                m = res.metrics or {}
                if key not in m:
                    continue
                val = _safe_float(m.get(key), default=np.nan)
                if not np.isfinite(val):
                    continue
                n          = float(res.num_examples)
                numerator += n * val
                denom     += n
            return float(numerator / denom) if denom > 0 else default

        srv_train_loss     = wavg("train_loss")
        srv_train_acc      = wavg("train_acc")
        srv_val_loss       = wavg("val_loss")
        srv_val_acc        = wavg("val_acc")
        srv_val_precision  = wavg("val_precision")
        srv_val_recall     = wavg("val_recall")
        srv_val_f1         = wavg("val_f1")
        srv_test_loss      = wavg("test_loss")
        srv_test_acc       = wavg("test_acc")
        srv_test_precision = wavg("test_precision")
        srv_test_recall    = wavg("test_recall")
        srv_test_f1        = wavg("test_f1")
        srv_avg_local_loss = wavg("avg_local_loss")

        early_stopped_count = 0
        banned_count        = 0

        # ── Per-client history ───────────────────────────────
        for client_proxy, fitres in results:
            cid = str(getattr(client_proxy, "cid", "?"))
            _init_client_history(cid)

            m = fitres.metrics or {}

            early_stopped = _safe_int(m.get("early_stopped", 0), 0)
            banned_client = _safe_int(m.get("banned_client",  0), 0)

            early_stopped_count += early_stopped
            banned_count        += banned_client

            h = FEDSEM_HIST["clients"][cid]

            h["round"].append(int(server_round))
            h["train_loss"].append(_safe_float(m.get("train_loss", 0.0)))
            h["train_acc"].append(_safe_float(m.get("train_acc", 0.0)))
            h["val_loss"].append(_safe_float(m.get("val_loss", 0.0)))
            h["val_acc"].append(_safe_float(m.get("val_acc", 0.0)))
            h["val_precision"].append(_safe_float(m.get("val_precision", 0.0)))
            h["val_recall"].append(_safe_float(m.get("val_recall", 0.0)))
            h["val_f1"].append(_safe_float(m.get("val_f1", 0.0)))
            h["val_cm_json"].append(_safe_json_or_none(m.get("val_cm_json", None)))
            h["test_loss"].append(_safe_float(m.get("test_loss", 0.0)))
            h["test_acc"].append(_safe_float(m.get("test_acc", 0.0)))
            h["test_precision"].append(_safe_float(m.get("test_precision", 0.0)))
            h["test_recall"].append(_safe_float(m.get("test_recall", 0.0)))
            h["test_f1"].append(_safe_float(m.get("test_f1", 0.0)))
            h["test_cm_json"].append(_safe_json_or_none(m.get("test_cm_json", None)))
            h["avg_local_loss"].append(_safe_float(m.get("avg_local_loss", 0.0)))
            h["n_train"].append(_safe_int(m.get("n_train", fitres.num_examples)))
            h["n_val"].append(_safe_int(m.get("n_val", 0)))
            h["n_test"].append(_safe_int(m.get("n_test", 0)))
            h["share_weight"].append(_safe_int(m.get("share_weight", fitres.num_examples)))
            h["tau"].append(_safe_int(m.get("tau", 0)))
            h["banned_client"].append(banned_client)
            h["loss_history_json"].append(_safe_json_or_none(m.get("loss_history_json", None)))
            h["train_loss_curve_json"].append(_safe_json_or_none(m.get("train_loss_curve_json", None)))
            h["val_loss_curve_json"].append(_safe_json_or_none(m.get("val_loss_curve_json", None)))
            h["early_stopped"].append(early_stopped)
            h["stopped_epoch"].append(_safe_int(m.get("stopped_epoch", -1), -1))
            h["best_val_loss"].append(_safe_float(m.get("best_val_loss", np.nan), np.nan))
            h["update_l2"].append(_safe_float(m.get("update_l2", m.get("delta_l2", 0.0))))
            h["has_protos"].append(_safe_int(m.get("has_protos", 0)))
            h["sep_score_client"].append(_safe_float(m.get("sep_score_client", 0.0)))
            h["proto_missing_0"].append(_safe_int(m.get("proto_missing_0", 0)))
            h["proto_missing_1"].append(_safe_int(m.get("proto_missing_1", 0)))
            h["proto_0_norm"].append(_safe_float(m.get("proto_0_norm", 0.0)))
            h["proto_1_norm"].append(_safe_float(m.get("proto_1_norm", 0.0)))
            h["proto_sep_cos"].append(_safe_float(m.get("proto_sep_cos", 0.0)))
            h["proto_sep_angle"].append(_safe_float(m.get("proto_sep_angle", 0.0)))
            h["proto_sep_weight"].append(_safe_float(m.get("proto_sep_weight", 0.0)))
            h["proto_sep_score"].append(_safe_float(m.get("proto_sep_score", 0.0)))
            h["proto_n0"].append(_safe_int(m.get("proto_n0", 0)))
            h["proto_n1"].append(_safe_int(m.get("proto_n1", 0)))
            h["proto_0_vec_json"].append(m.get("proto_0_vec_json", None))
            h["proto_1_vec_json"].append(m.get("proto_1_vec_json", None))

        # ── Server history ───────────────────────────────────
        FEDSEM_HIST["server"]["round"].append(int(server_round))
        FEDSEM_HIST["server"]["train_loss_w"].append(float(srv_train_loss))
        FEDSEM_HIST["server"]["train_acc_w"].append(float(srv_train_acc))
        FEDSEM_HIST["server"]["val_loss_w"].append(float(srv_val_loss))
        FEDSEM_HIST["server"]["val_acc_w"].append(float(srv_val_acc))
        FEDSEM_HIST["server"]["val_precision_w"].append(float(srv_val_precision))
        FEDSEM_HIST["server"]["val_recall_w"].append(float(srv_val_recall))
        FEDSEM_HIST["server"]["val_f1_w"].append(float(srv_val_f1))
        FEDSEM_HIST["server"]["test_loss_w"].append(float(srv_test_loss))
        FEDSEM_HIST["server"]["test_acc_w"].append(float(srv_test_acc))
        FEDSEM_HIST["server"]["test_precision_w"].append(float(srv_test_precision))
        FEDSEM_HIST["server"]["test_recall_w"].append(float(srv_test_recall))
        FEDSEM_HIST["server"]["test_f1_w"].append(float(srv_test_f1))
        FEDSEM_HIST["server"]["avg_local_loss_w"].append(float(srv_avg_local_loss))
        FEDSEM_HIST["server"]["early_stopped_count"].append(int(early_stopped_count))
        FEDSEM_HIST["server"]["active_clients"].append(int(len(results)))
        FEDSEM_HIST["server"]["banned_clients"].append(int(banned_count))
        FEDSEM_HIST["server"]["failures"].append(int(len(failures)))

        # Store latest aggregated params
        try:
            FEDSEM_GLOBAL_FINAL["params"] = parameters_to_ndarrays(new_params)
        except Exception:
            FEDSEM_GLOBAL_FINAL["params"] = None

        # ── Print summary ────────────────────────────────────
        print(
            f"[Server][Round {server_round}] === {self.title} === "
            f"(clients={len(results)} failures={len(failures)})"
        )
        print(
            f"[Server Summary | Round {server_round}] "
            f"TrLoss={srv_train_loss:.4f} TrAcc={srv_train_acc:.4f} | "
            f"VaLoss={srv_val_loss:.4f} VaAcc={srv_val_acc:.4f} VaF1={srv_val_f1:.4f} | "
            f"TeLoss={srv_test_loss:.4f} TeAcc={srv_test_acc:.4f} TeF1={srv_test_f1:.4f} | "
            f"ES={early_stopped_count}/{len(results)}"
        )

        if self.print_client_table:
            header = (
                f"{'CID':>3} | {'n_tr':>6} {'n_va':>5} {'n_te':>5} | "
                f"{'TrL':>6} {'TrA':>5} {'VaL':>6} {'VaA':>5} {'VaF1':>5} "
                f"{'TeL':>6} {'TeA':>5} {'TeF1':>5} | "
                f"{'AvgL':>6} {'||Δ||':>10} | "
                f"{'ES':>2} {'Stop':>4} {'BestVL':>7} | "
                f"{'HP':>2} {'P0m':>3} {'P1m':>3} {'Sep':>6}"
            )
            print(header)
            print("-" * len(header))

            for client_proxy, fitres in results:
                cid = str(getattr(client_proxy, "cid", "?"))
                m   = fitres.metrics or {}

                n_tr    = _safe_int(m.get("n_train", fitres.num_examples))
                n_va    = _safe_int(m.get("n_val", 0))
                n_te    = _safe_int(m.get("n_test", 0))
                trl     = _safe_float(m.get("train_loss", 0.0))
                tra     = _safe_float(m.get("train_acc",  0.0))
                val     = _safe_float(m.get("val_loss",   0.0))
                vaa     = _safe_float(m.get("val_acc",    0.0))
                vf1     = _safe_float(m.get("val_f1",     0.0))
                tel     = _safe_float(m.get("test_loss",  0.0))
                tea     = _safe_float(m.get("test_acc",   0.0))
                tf1     = _safe_float(m.get("test_f1",    0.0))
                avgl    = _safe_float(m.get("avg_local_loss", 0.0))
                dl2     = _safe_float(m.get("update_l2", m.get("delta_l2", 0.0)))
                es      = _safe_int(m.get("early_stopped", 0))
                stop_ep = _safe_int(m.get("stopped_epoch", -1), -1)
                best_vl = _safe_float(m.get("best_val_loss", np.nan), np.nan)
                hp      = _safe_int(m.get("has_protos", 0))
                p0m     = _safe_int(m.get("proto_missing_0", 0))
                p1m     = _safe_int(m.get("proto_missing_1", 0))
                sep     = _safe_float(m.get("proto_sep_score", 0.0))

                best_vl_str = f"{best_vl:7.3f}" if np.isfinite(best_vl) else f"{'nan':>7}"

                print(
                    f"{cid:>3} | "
                    f"{n_tr:6d} {n_va:5d} {n_te:5d} | "
                    f"{trl:6.3f} {tra:5.3f} {val:6.3f} {vaa:5.3f} {vf1:5.3f} "
                    f"{tel:6.3f} {tea:5.3f} {tf1:5.3f} | "
                    f"{avgl:6.3f} {dl2:10.3e} | "
                    f"{es:2d} {stop_ep:4d} {best_vl_str} | "
                    f"{hp:2d} {p0m:3d} {p1m:3d} {sep:6.3f}"
                )

        metrics_agg = {
            "train_loss_w":        float(srv_train_loss),
            "train_acc_w":         float(srv_train_acc),
            "val_loss_w":          float(srv_val_loss),
            "val_acc_w":           float(srv_val_acc),
            "val_precision_w":     float(srv_val_precision),
            "val_recall_w":        float(srv_val_recall),
            "val_f1_w":            float(srv_val_f1),
            "test_loss_w":         float(srv_test_loss),
            "test_acc_w":          float(srv_test_acc),
            "test_precision_w":    float(srv_test_precision),
            "test_recall_w":       float(srv_test_recall),
            "test_f1_w":           float(srv_test_f1),
            "avg_local_loss_w":    float(srv_avg_local_loss),
            "early_stopped_count": int(early_stopped_count),
            "active_clients":      int(len(results)),
            "banned_clients":      int(banned_count),
            "failures":            int(len(failures)),
        }

        return new_params, metrics_agg


# ============================================================
# Save History Helper
# ============================================================

def save_fedsem_history(path: str = "fedsem_server_history.json") -> None:
    """Save FEDSEM_HIST as JSON for graph drawing."""
    with open(path, "w", encoding="utf-8") as f:
        json.dump(FEDSEM_HIST, f, indent=4, ensure_ascii=False)
    print(f"[History] Saved FEDSEM_HIST to: {path}")
```

```python
import json
from typing import Dict, List, Tuple, Optional, Set, Any
from collections import defaultdict

import numpy as np
import pandas as pd

import flwr as fl
from flwr.common import Parameters, Scalar, FitRes, ndarrays_to_parameters, parameters_to_ndarrays
from flwr.server.strategy import FedAvg

import torch
from torch.utils.data import Dataset, DataLoader


# ============================================================
# TrustDFDataset + loader
# ============================================================

class TrustDFDataset(Dataset):
    def __init__(self, df: pd.DataFrame, tokenizer, max_len: int = 48):
        self.texts     = df["review_clean"].astype(str).tolist()
        self.labels    = df["y"].astype(int).tolist()
        self.tokenizer = tokenizer
        self.max_len   = max_len

    def __len__(self) -> int:
        return len(self.texts)

    def __getitem__(self, idx: int) -> Dict[str, torch.Tensor]:
        enc  = self.tokenizer(
            self.texts[idx],
            truncation=True,
            padding="max_length",
            max_length=self.max_len,
            return_tensors="pt",
        )
        item = {k: v.squeeze(0) for k, v in enc.items()}
        item["labels"] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item


def make_trust_loader(
    df_trust: pd.DataFrame,
    tokenizer,
    batch_size: int = 32,
    max_len: int = 48,
    shuffle: bool = False,
) -> DataLoader:
    yvals = set(df_trust["y"].dropna().astype(int).unique().tolist())
    assert yvals.issubset({0, 1}), \
        f"Trust labels must be binary {{0,1}}, got {yvals}"
    assert 0 in yvals and 1 in yvals, \
        "Trust set must contain BOTH classes 0 and 1"
    ds = TrustDFDataset(df_trust, tokenizer, max_len=max_len)
    return DataLoader(ds, batch_size=batch_size, shuffle=shuffle)


# ============================================================
# Robust math helpers
# ============================================================

def geometric_median(
    X: np.ndarray,
    weights: Optional[np.ndarray] = None,
    eps: float = 1e-6,
    max_iter: int = 200,
) -> np.ndarray:
    X = np.asarray(X, dtype=np.float64)
    m, _ = X.shape
    if m == 1:
        return X[0].astype(np.float32)
    if weights is None:
        w = np.ones(m, dtype=np.float64)
    else:
        w = np.asarray(weights, dtype=np.float64)
        w = np.clip(w, 1e-12, None)
    y = (w[:, None] * X).sum(axis=0) / w.sum()
    for _ in range(max_iter):
        diff = X - y[None, :]
        dist = np.linalg.norm(diff, axis=1)
        if np.any(dist < eps):
            y = X[np.argmin(dist)]
            break
        inv   = w / np.clip(dist, eps, None)
        y_new = (inv[:, None] * X).sum(axis=0) / inv.sum()
        if np.linalg.norm(y_new - y) < eps:
            y = y_new
            break
        y = y_new
    return y.astype(np.float32)


def l2n(v: np.ndarray, eps: float = 1e-12) -> np.ndarray:
    v = v.astype(np.float32)
    n = float(np.linalg.norm(v))
    return v / (n + eps)


def mad(x: np.ndarray, eps: float = 1e-8) -> float:
    x   = np.asarray(x, dtype=np.float32)
    med = float(np.median(x)) if x.size else 0.0
    return float(np.median(np.abs(x - med)) + eps)


def robust_z(x: np.ndarray, eps: float = 1e-8) -> np.ndarray:
    x    = np.asarray(x, dtype=np.float32)
    med  = float(np.median(x)) if x.size else 0.0
    madv = float(np.median(np.abs(x - med))) + eps
    return (x - med) / (1.4826 * madv)


def sigmoid(x: np.ndarray) -> np.ndarray:
    x = np.asarray(x, dtype=np.float32)
    return 1.0 / (1.0 + np.exp(-x))


def mad_cv(x: np.ndarray, eps: float = 1e-12) -> float:
    x = np.asarray(x, dtype=np.float32)
    if x.size < 2:
        return 0.0
    med  = float(np.median(x))
    madv = float(np.median(np.abs(x - med)))
    return float((1.4826 * madv) / (abs(med) + eps))


def winsorized_max_thr(
        hist: List[float],
        floor: float = 0.07,
        alpha: float = 1.5) -> float:
    h = np.asarray(hist, dtype=np.float32)
    if h.size == 0:
        return float(floor)
    med    = float(np.median(h))
    madv   = float(np.median(np.abs(h - med)))
    cap    = med + alpha * madv
    h_clip = np.minimum(h, cap)
    return float(max(floor, float(np.max(h_clip))))


# ============================================================
# KMeans + BIC (k in {1,2}) — 1D guarded
# ============================================================

def kmeans_bic_hard_split_1d_guarded(
    x: np.ndarray,
    seed: int = 0,
    n_init: int = 50,
    min_cluster_size: int = 2,
) -> Tuple[int, np.ndarray, np.ndarray, np.ndarray, Dict[str, Any]]:
    from sklearn.cluster import KMeans

    x0 = np.asarray(x, dtype=np.float32).reshape(-1)
    X  = x0.reshape(-1, 1)
    n  = int(X.shape[0])
    d  = 1

    dbg: Dict[str, Any] = {
        "n": n, "min_cluster_size": int(min_cluster_size),
        "sse1": None, "sse2": None, "bic1": None, "bic2": None,
        "c_low": None, "c_high": None, "forced_k1": None,
        "centers": None, "choose_k_raw": None, "choose_k_final": None,
    }

    if n < 2:
        labels    = np.zeros(n, dtype=int)
        centers   = np.array([float(np.mean(x0))], dtype=np.float32)
        order_map = np.array([0], dtype=int)
        dbg["centers"] = centers.tolist()
        dbg["choose_k_raw"] = 1
        dbg["choose_k_final"] = 1
        return 1, labels, centers, order_map, dbg

    def fit_k(k):
        km = KMeans(n_clusters=k, random_state=seed, n_init=n_init)
        km.fit(X)
        return (km.labels_.astype(int),
                km.cluster_centers_.reshape(-1).astype(np.float32),
                float(km.inertia_))

    def bic_from_sse(sse, k):
        denom  = max(1, (n - k))
        sigma2 = max(1e-12, sse / denom)
        ll     = (-0.5 * n * d * np.log(2.0 * np.pi * sigma2)
                  - 0.5 * sse / sigma2)
        return float(-2.0 * ll + (k * d + 1) * np.log(n))

    lab1, cen1, sse1 = fit_k(1)
    lab2, cen2, sse2 = fit_k(2)
    bic1 = bic_from_sse(sse1, 1)
    bic2 = bic_from_sse(sse2, 2)
    dbg.update({"sse1": sse1, "sse2": sse2, "bic1": bic1, "bic2": bic2})

    order2   = np.argsort(cen2)
    low_lab  = int(order2[0])
    high_lab = int(order2[1])
    c_low    = int(np.sum(lab2 == low_lab))
    c_high   = int(np.sum(lab2 == high_lab))
    dbg["c_low"], dbg["c_high"] = c_low, c_high

    choose_k = 1 if bic1 <= bic2 else 2
    dbg["choose_k_raw"] = int(choose_k)

    if choose_k == 2 and min(c_low, c_high) < int(min_cluster_size):
        choose_k = 1
        dbg["forced_k1"] = (
            f"tiny_cluster(min={min(c_low,c_high)}<{min_cluster_size})")

    labels  = lab1 if choose_k == 1 else lab2
    centers = cen1 if choose_k == 1 else cen2
    order   = np.argsort(centers)
    order_map = np.zeros_like(order)
    for r, lab in enumerate(order):
        order_map[lab] = r
    centers_sorted = centers[order]

    dbg["centers"]        = centers_sorted.tolist()
    dbg["choose_k_final"] = int(choose_k)
    return int(choose_k), labels, centers_sorted, order_map, dbg


# ============================================================
# Directional outlier-aware BIC (1D)
# ============================================================

def kmeans_bic_outlier_directional_1d(
    x: np.ndarray,
    seed: int = 0,
    n_init: int = 50,
    probe_min_cluster_size: int = 1,
    core_min_cluster_size: int = 1,
    keep_low_singleton_if_core_k1: bool = True,
) -> Tuple[int, np.ndarray, np.ndarray, np.ndarray, Dict[str, Any]]:
    x0 = np.asarray(x, dtype=np.float32).reshape(-1)
    n  = int(x0.size)

    dbg: Dict[str, Any] = {
        "n": n,
        "probe_min_cluster_size": int(probe_min_cluster_size),
        "core_min_cluster_size":  int(core_min_cluster_size),
        "outlier_idx":            [],
        "outlier_side":           None,
        "probe_dbg":              None,
        "core_dbg":               None,
        "centers_final":          None,
        "forced_k2_reason":       None,
        "outlier_only_suspicious": False,
        "core_idx":               [],
        "core_low_idx":           [],
        "core_high_idx":          [],
        "core_k":                 None,
    }

    if n < 2:
        labels    = np.zeros(n, dtype=int)
        centers   = np.array([float(np.mean(x0))], dtype=np.float32)
        order_map = np.array([0], dtype=int)
        dbg["centers_final"] = centers.tolist()
        return 1, labels, centers, order_map, dbg

    kP, labP, centersP_sorted, order_mapP, probe_dbg = \
        kmeans_bic_hard_split_1d_guarded(
            x0, seed=seed, n_init=n_init,
            min_cluster_size=probe_min_cluster_size)
    dbg["probe_dbg"] = probe_dbg

    choose_raw = int((probe_dbg or {}).get("choose_k_raw", 1))
    c_low  = int((probe_dbg or {}).get("c_low",  0) or 0)
    c_high = int((probe_dbg or {}).get("c_high", 0) or 0)

    outlier_idx:  List[int] = []
    outlier_side: Optional[str] = None

    if choose_raw == 2 and min(c_low, c_high) == 1:
        unique, counts = np.unique(labP, return_counts=True)
        singleton_labs = [
            int(u) for u, cnt in zip(unique, counts)
            if int(cnt) == 1]
        if len(singleton_labs) == 1:
            singleton_lab = singleton_labs[0]
            outlier_idx   = [
                int(i) for i in
                np.where(labP == singleton_lab)[0].tolist()]
            rank         = int(order_mapP[int(singleton_lab)])
            outlier_side = "LOW" if rank == 0 else "HIGH"

    dbg["outlier_idx"]  = outlier_idx
    dbg["outlier_side"] = outlier_side

    if len(outlier_idx) == 0 or n - len(outlier_idx) < 2:
        k, labels, centers_sorted, order_map, core_dbg = \
            kmeans_bic_hard_split_1d_guarded(
                x0, seed=seed, n_init=n_init,
                min_cluster_size=core_min_cluster_size)
        dbg["core_dbg"]      = core_dbg
        dbg["centers_final"] = (
            centers_sorted.tolist()
            if centers_sorted is not None else None)
        return k, labels, centers_sorted, order_map, dbg

    core_set = set(outlier_idx)
    core_idx = [i for i in range(n) if i not in core_set]
    x_core   = x0[core_idx]
    dbg["core_idx"] = [int(i) for i in core_idx]

    kC, labC, centersC_sorted, order_mapC, core_dbg = \
        kmeans_bic_hard_split_1d_guarded(
            x_core, seed=seed, n_init=n_init,
            min_cluster_size=core_min_cluster_size)
    dbg["core_dbg"] = core_dbg
    dbg["core_k"]   = int(kC)

    labels_full = np.full(n, -1, dtype=int)
    labels_full[core_idx] = labC

    if kC == 1:
        if outlier_side == "HIGH":
            labels_full[outlier_idx] = 0
            centers_full   = np.array(
                [float(np.mean(x0))], dtype=np.float32)
            order_map_full = np.array([0], dtype=int)
            dbg["centers_final"] = centers_full.tolist()
            return 1, labels_full, centers_full, order_map_full, dbg

        core_center = float(np.mean(x0[core_idx]))
        out_center  = float(np.mean(x0[outlier_idx]))

        if out_center <= core_center:
            labels_full[core_idx]    = 1
            labels_full[outlier_idx] = 0
            centers_sorted = np.array(
                [out_center, core_center], dtype=np.float32)
        else:
            labels_full[core_idx]    = 0
            labels_full[outlier_idx] = 1
            centers_sorted = np.array(
                [core_center, out_center], dtype=np.float32)

        order_map = np.array([0, 1], dtype=int)
        dbg["forced_k2_reason"]        = \
            "LOW_singleton_core_k1_ALWAYS_FORCE_K2"
        dbg["outlier_only_suspicious"] = True
        dbg["centers_final"]           = centers_sorted.tolist()
        return 2, labels_full, centers_sorted, order_map, dbg

    low_c  = float(centersC_sorted[0])
    high_c = float(centersC_sorted[1])

    unique_labs = np.unique(labC).astype(int)
    rank_to_lab = np.full(2, -1, dtype=int)
    for lab in unique_labs:
        r = int(order_mapC[lab])
        if r in (0, 1):
            rank_to_lab[r] = int(lab)
    if np.any(rank_to_lab < 0):
        rank_to_lab = np.array([0, 1], dtype=int)

    core_low_lab  = int(rank_to_lab[0])
    core_high_lab = int(rank_to_lab[1])
    dbg["core_low_idx"]  = [
        int(core_idx[j]) for j in range(len(core_idx))
        if int(labC[j]) == core_low_lab]
    dbg["core_high_idx"] = [
        int(core_idx[j]) for j in range(len(core_idx))
        if int(labC[j]) == core_high_lab]

    for oi in outlier_idx:
        xi   = float(x0[oi])
        rank = 0 if abs(xi - low_c) <= abs(xi - high_c) else 1
        labels_full[oi] = int(rank_to_lab[rank])

    centers = []
    for lab in [0, 1]:
        pts = x0[labels_full == lab]
        centers.append(
            float(np.mean(pts)) if pts.size else float("nan"))
    centers = np.asarray(centers, dtype=np.float32)

    order = np.argsort(centers)
    centers_sorted = centers[order]
    order_map = np.zeros_like(order)
    for r, lab in enumerate(order):
        order_map[lab] = r

    if outlier_side == "LOW":
        dbg["outlier_only_suspicious"] = True

    dbg["centers_final"] = centers_sorted.tolist()
    return 2, labels_full, centers_sorted, order_map, dbg


# ============================================================
# FedSemDefenceStrategy v3.8.0
# ============================================================

class FedSemDefenceStrategy(FedAvg):
    def __init__(self,
        trust_clip_min: float = 0.0,
        trust_clip_max: float = 1.0,
        ban_after_flags: int = 3,
        ref_alpha_mad: float = 1.0,
        score_cv_eps: float = 1e-12,
        score_cv_floor: float = 0.07,
        score_cv_win_alpha: float = 1.5,
        geo_weight_by_examples: bool = True,
        missing_proto_strike_after: int = 2,
        use_trust_for_flagging: bool = True,
        kmeans_random_state: int = 0,
        kmeans_n_init: int = 50,
        kmeans_probe_min_cluster_size: int = 1,
        kmeans_core_min_cluster_size: int = 1,
        use_outlier_bic: bool = True,
        keep_low_singleton_if_core_k1: bool = True,
        outlier_apply_to_trust: bool = True,
        outlier_apply_to_dsep: bool = True,
        use_cv_gate: bool = True,
        cv_eps: float = 1e-12,
        cv_floor: float = 0.07,
        cv_win_alpha: float = 1.5,
        trust_lambda_swap: float = 0.5,
        df_trust=None,
        trust_tokenizer=None,
        trust_server_model=None,
        trust_device=None,
        trust_batch_size: int = 32,
        trust_max_len: int = 48,
        trust_mom_K: int = 5,
        trust_mom_seed: int = 42,
        num_labels: int = 2,
        title: str = (
            "FedSemDefence v3.8.0 "
            "(dual clustering + directional outlier BIC + "
            "MAD validation + adaptive ScoreRef FlagGate + "
            "singleton cancel via lowref intersection)"
        ),
        **kwargs,
    ):
        super().__init__(**kwargs)

        self.min_trust               = float(trust_clip_min)
        self.max_trust               = float(trust_clip_max)
        self.ban_after_flags         = int(ban_after_flags)
        self.ref_alpha_mad           = float(ref_alpha_mad)
        self.score_cv_eps            = float(score_cv_eps)
        self.score_cv_floor          = float(score_cv_floor)
        self.score_cv_win_alpha      = float(score_cv_win_alpha)
        self.score_cv_hist: List[float] = []
        self.geo_weight_by_examples  = bool(geo_weight_by_examples)
        self.missing_proto_strike_after = int(missing_proto_strike_after)
        self.use_trust_for_flagging  = bool(use_trust_for_flagging)
        self.kmeans_random_state     = int(kmeans_random_state)
        self.kmeans_n_init           = int(kmeans_n_init)
        self.kmeans_probe_min_cluster_size = \
            int(kmeans_probe_min_cluster_size)
        self.kmeans_core_min_cluster_size  = \
            int(kmeans_core_min_cluster_size)
        self.use_outlier_bic         = bool(use_outlier_bic)
        self.keep_low_singleton_if_core_k1 = \
            bool(keep_low_singleton_if_core_k1)
        self.outlier_apply_to_trust  = bool(outlier_apply_to_trust)
        self.outlier_apply_to_dsep   = bool(outlier_apply_to_dsep)
        self.use_cv_gate             = bool(use_cv_gate)
        self.cv_eps                  = float(cv_eps)
        self.cv_floor                = float(cv_floor)
        self.cv_win_alpha            = float(cv_win_alpha)
        self.cv_hist: List[float]    = []
        self.trust_lambda_swap       = float(trust_lambda_swap)
        self.title                   = str(title)

        self.client_flag_count:   Dict[int, int] = defaultdict(int)
        self.missing_proto_count: Dict[int, int] = defaultdict(int)
        self.banned_clients:      Set[int]        = set()
        self.server_history:      Dict[int, Dict[str, Any]] = {}
        self._prev_global_proto:  Optional[
            Tuple[np.ndarray, np.ndarray]] = None

        # ✅ NEW — adaptive ScoreRef FlagGate history
        # Stores minimum ScoreRef of confirmed honest clients per round
        # Used to compute adaptive threshold for flag confirmation
        self._client_scoreref_hist: Dict[int, List[float]] = defaultdict(list)

        print("\n[Server] ✅ FedSemDefence v3.8.0 init")
        print(f"  trust_lambda_swap       = {self.trust_lambda_swap}")
        print("  Dual clustering         = "
              "Trust AND Δsep separately → intersection")
        print("  Outlier handling        = "
              "directional BIC (LOW/HIGH singleton)")
        print("  LOW singleton (kC=2)    = "
              "MAD validation (Trust + Δsep)")
        print("  Singleton cancel rule   = "
              "if final=1 AND NOT in lowref → cancel")
        print("  B-rule (cv_brule)       = REMOVED")
        print("  Rule 8 (ScoreRef fallback) = REMOVED")
        print("  FlagGate               = "
              "adaptive ScoreRef min history + swap>0")

        self.mu_ref: Optional[Dict[int, np.ndarray]] = None
        if df_trust is not None:
            assert trust_tokenizer    is not None
            assert trust_server_model is not None
            assert trust_device       is not None

            trust_loader = make_trust_loader(
                df_trust   = df_trust,
                tokenizer  = trust_tokenizer,
                batch_size = trust_batch_size,
                max_len    = trust_max_len,
                shuffle    = False,
            )

            print("[Server] Computing reference prototypes "
                  "from df_trust...")
            trust_server_model.eval()
            trust_server_model.to(trust_device)

            if hasattr(trust_server_model, "config"):
                trust_server_model.config.output_hidden_states = True

            class_embeddings = defaultdict(list)

            with torch.no_grad():
                for batch in trust_loader:
                    input_ids      = batch["input_ids"].to(
                        trust_device)
                    attention_mask = batch["attention_mask"].to(
                        trust_device)
                    labels         = batch["labels"].cpu().numpy()

                    fwd = {
                        "input_ids":            input_ids,
                        "attention_mask":       attention_mask,
                        "output_hidden_states": True,
                        "return_dict":          True,
                    }
                    if "token_type_ids" in batch:
                        fwd["token_type_ids"] = batch[
                            "token_type_ids"].to(trust_device)

                    outputs = trust_server_model(**fwd)

                    if hasattr(outputs, "hidden_states") and \
                            outputs.hidden_states is not None:
                        hidden_states = outputs.hidden_states[-1]
                    else:
                        hidden_states = outputs.last_hidden_state

                    cls_embeddings = \
                        hidden_states[:, 0, :].cpu().numpy()
                    norms = np.linalg.norm(
                    cls_embeddings, axis=1, keepdims=True)
                    cls_embeddings = cls_embeddings / (norms + 1e-12)

                    for sample_emb, sample_label in zip(
                            cls_embeddings, labels):
                        class_embeddings[
                            int(sample_label)].append(sample_emb)

            

            ref_extracted = {}
            for class_idx in range(num_labels):
                if class_embeddings[class_idx]:
                    embs_arr = np.stack(
                        class_embeddings[class_idx], axis=0)
                    ref_extracted[class_idx] = \
                        geometric_median(embs_arr)
                else:
                    raise ValueError(
                        f"Class {class_idx} has no samples "
                        f"in df_trust!")

            assert ref_extracted[0] is not None and \
                   ref_extracted[1] is not None

            self.mu_ref = {
                0: ref_extracted[0].astype(np.float32),
                1: ref_extracted[1].astype(np.float32),
            }
            print("[Server] ✅ Reference prototypes built "
                  "from df_trust")

            self._server_trust_loader = make_trust_loader(
                df_trust   = df_trust,
                tokenizer  = trust_tokenizer,
                batch_size = trust_batch_size,
                max_len    = trust_max_len,
                shuffle    = False,
            )
            self._trust_device = trust_device
            _temp_model      = init_banglabert_classifier(
                num_labels=num_labels)
            _temp_state      = get_trainable_state_dict(_temp_model)
            self._param_keys = list(_temp_state.keys())
            del _temp_model
            torch.cuda.empty_cache()
                        
            
            # ── Cache proto model — built once, reused every round ──
            self._proto_model = init_banglabert_classifier(
                num_labels=num_labels)
            self._proto_model.to(trust_device)
            self._proto_model.eval()
            print("[Server] ✅ Proto model cached — reused every round")
            # ────────────────────────────────────────────────────────
            print("[Server] ✅ Server-side proto computation ready")

        else:
            print("[Server] ⚠️ df_trust not provided.")
            self._server_trust_loader = None
            self._trust_device        = None
            self._param_keys          = None

    @staticmethod
    def _cid_from(cp, fr=None) -> int:
        if fr is not None and fr.metrics is not None \
                and "cid" in fr.metrics:
            try:
                return int(fr.metrics["cid"])
            except Exception:
                pass
        try:
            return int(getattr(cp, "cid"))
        except Exception:
            return abs(hash(
                str(getattr(cp, "cid", "unknown")))) % (10**9)

    @staticmethod
    def _cos(a, b) -> float:
        a = np.asarray(a, dtype=np.float32)
        b = np.asarray(b, dtype=np.float32)
        if not (np.all(np.isfinite(a)) and
                np.all(np.isfinite(b))):
            return 0.0
        na = float(np.linalg.norm(a))
        nb = float(np.linalg.norm(b))
        denom = na * nb + 1e-8
        if denom <= 0:
            return 0.0
        v = float(np.dot(a, b) / denom)
        return 0.0 if not np.isfinite(v) else v

    def _cv_score(self, x) -> float:
        x = np.asarray(x, dtype=np.float32)
        if x.size < 2:
            return 0.0
        mu  = float(np.mean(x))
        sig = float(np.std(x))
        if abs(mu) < self.score_cv_eps:
            return 0.0
        return float(sig / (abs(mu) + self.score_cv_eps))

    def _wdata(self, cid_list, n_by_cid):
        N = np.asarray(
            [max(0, int(n_by_cid.get(c, 0))) for c in cid_list],
            dtype=np.float32)
        s = float(np.sum(N))
        if s <= 0:
            return np.ones(len(cid_list), dtype=np.float32)
        w = 1.0 - (N / (s + 1e-12))
        return np.clip(w, 0.0, 1.0).astype(np.float32)

    def _build_global_proto_geomed(
        self, cid_list, mu0_list, mu1_list,
        n_by_cid, selected):
        sel = set(selected)
        Z0, Z1 = [], []
        for i, cid in enumerate(cid_list):
            if cid not in sel:
                continue
            m0, m1 = mu0_list[i], mu1_list[i]
            if m0 is not None:
                Z0.append(m0.astype(np.float32))
            if m1 is not None:
                Z1.append(m1.astype(np.float32))
        g0 = g1 = None
        if Z0:
            g0 = mom_prototype(
                np.stack(Z0, 0),
                K=min(MOM_K, len(Z0)))
        if Z1:
            g1 = mom_prototype(
                np.stack(Z1, 0),
                K=min(MOM_K, len(Z1)))
        return g0, g1

    def _dynamic_ref_split(
            self, cid_list, score_ref_split, score_ref_cv_raw):
        s = np.asarray(score_ref_split, dtype=np.float32)
        n = len(cid_list)
        if n == 0:
            return ([], [], 0.0, 0.0, 0.0, 0.0,
                    self.score_cv_floor, "EMPTY")
        forced_low = [
            cid_list[i] for i in range(n)
            if float(s[i]) < 0.0]
        remaining  = [
            cid_list[i] for i in range(n)
            if float(s[i]) >= 0.0]
        s_rem = np.asarray(
            [s[i] for i in range(n) if float(s[i]) >= 0.0],
            dtype=np.float32)
        s_med   = float(np.median(s_rem)) if s_rem.size else 0.0
        s_range = float(
            s_rem.max() - s_rem.min()) if s_rem.size else 0.0
        cv_ref  = float(score_ref_cv_raw)
        score_cv_thr_round = winsorized_max_thr(
            self.score_cv_hist,
            floor=self.score_cv_floor,
            alpha=self.score_cv_win_alpha)
        if len(remaining) == 0:
            return ([], forced_low[:], 0.0, s_range, s_med,
                    cv_ref, score_cv_thr_round, "ALL_NEG")
        if cv_ref <= score_cv_thr_round:
            return (remaining[:], forced_low[:], s_med, s_range,
                    s_med, cv_ref, score_cv_thr_round, "NO_SPLIT_CV")
        s_mad        = mad(s_rem)
        thr          = s_med - self.ref_alpha_mad * s_mad
        low_from_thr = [
            remaining[i] for i in range(len(remaining))
            if float(s_rem[i]) < thr]
        high = [c for c in remaining if c not in low_from_thr]
        low  = forced_low + low_from_thr
        if len(high) == 0:
            high = remaining[:]
            low  = forced_low[:]
        return (high, low, thr, s_range, s_med, cv_ref,
                score_cv_thr_round, "MAD_SPLIT")

    def configure_evaluate(
        self,
        server_round: int,
        parameters: fl.common.Parameters,
        client_manager: fl.server.client_manager.ClientManager,
    ):
        # Client-side evaluation disabled.
        # Global model evaluated centrally via evaluate_fn.
        return []

    def configure_fit(
        self, server_round, parameters, client_manager):
        cfg      = super().configure_fit(
            server_round, parameters, client_manager)
        filtered = []
        for cp, fit_ins in cfg:
            cid = self._cid_from(cp, None)
            if cid in self.banned_clients:
                continue
            c = dict(fit_ins.config) if fit_ins.config else {}
            c["server_round"] = int(server_round)
            c["epochs"]       = LOCAL_EPOCHS

            filtered.append(
                (cp, fl.common.FitIns(fit_ins.parameters, c)))
        print(f"\n==================== FedSem Round "
              f"{server_round} ====================")
        if self.banned_clients:
            print(f"[Server] BANNED: {sorted(self.banned_clients)}")
        return filtered

    def aggregate_fit(self, rnd: int, results, failures):
        if not results:
            return None, {}
        results = [
            (cp, fr) for cp, fr in results
            if self._cid_from(cp, fr) not in self.banned_clients]
        if not results:
            return None, {
                "banned_count": int(len(self.banned_clients))}

        # ============================================================
        # Collect client results
        # ============================================================
        cid_list, dl2_list, mu0_list, mu1_list = [], [], [], []
        n_by_cid, fit_by_cid = {}, {}
        n0_by_cid, n1_by_cid, perf_by_cid = {}, {}, {}
        invalid_proto: Set[int] = set()

        for cp, fr in results:
            m   = fr.metrics or {}
            cid = self._cid_from(cp, fr)
            cid_list.append(cid)
            fit_by_cid[cid] = fr
            n = int(getattr(fr, "num_examples", 0) or 0)
            n_by_cid[cid] = n if n > 0 else 0
            dl2_list.append(float(
                m.get("update_l2", m.get("delta_l2", 0.0)) or 0.0))
            n0_by_cid[cid] = int(m.get(
                "proto_n0", m.get(
                    "n_class0_train", m.get("n0", 0))) or 0)
            n1_by_cid[cid] = int(m.get(
                "proto_n1", m.get(
                    "n_class1_train", m.get("n1", 0))) or 0)
            perf_by_cid[cid] = {
                "train_loss": float(m.get("train_loss", np.nan)),
                "train_acc":  float(m.get("train_acc",  np.nan)),
                "val_loss":   float(m.get("val_loss",   np.nan)),
                "val_acc":    float(m.get("val_acc",    np.nan)),
                "test_loss":  float(m.get("test_loss",  np.nan)),
                "test_acc":   float(m.get("test_acc",   np.nan)),
            }

            # ✅ Server-side prototype computation
            mu0 = mu1 = None
            if self._server_trust_loader is not None and \
                    self._param_keys is not None:
                try:
                    client_ndarrays = parameters_to_ndarrays(
                        fr.parameters)
                    temp_state = {
                        k: torch.tensor(v, device=self._trust_device)
                        for k, v in zip(
                            self._param_keys, client_ndarrays)
                    }
                    apply_trainable_state_dict(
                        self._proto_model, temp_state)
                    self._proto_model.eval()
                    class_embs = {0: [], 1: []}
                    with torch.no_grad():
                        for batch in self._server_trust_loader:
                            ids  = batch["input_ids"].to(
                                self._trust_device)
                            mask = batch["attention_mask"].to(
                                self._trust_device)
                            lbls = batch["labels"].cpu().numpy()
                            fwd  = {
                                "input_ids":            ids,
                                "attention_mask":       mask,
                                "output_hidden_states": True,
                                "return_dict":          True,
                            }
                            if "token_type_ids" in batch:
                                fwd["token_type_ids"] = batch[
                                    "token_type_ids"].to(
                                    self._trust_device)
                            outputs = self._proto_model(**fwd)
                            cls_emb = \
                                outputs.hidden_states[-1][:, 0, :]
                            cls_emb = torch.nn.functional.normalize(
                                        cls_emb, p=2, dim=1)
                            for emb, lbl in zip(
                                    cls_emb.cpu().numpy(), lbls):
                                class_embs[int(lbl)].append(emb)

                    if class_embs[0] and class_embs[1]:
                        mu0 = mom_prototype(
                            np.stack(
                                class_embs[0]).astype(np.float32),
                            K=MOM_K, seed=SEED)
                        mu1 = mom_prototype(
                            np.stack(
                                class_embs[1]).astype(np.float32),
                            K=MOM_K, seed=SEED+1)
                        print(f"  [ServerProto CID {cid}] ✅ "
                              f"n0={len(class_embs[0])} "
                              f"n1={len(class_embs[1])}")
                    else:
                        print(f"  [ServerProto CID {cid}] ❌ "
                              f"missing class")
                        invalid_proto.add(cid)


                except Exception as e:
                    print(f"  [ServerProto CID {cid}] ❌ "
                          f"error: {e}")
                    invalid_proto.add(cid)
                    mu0 = mu1 = None
            else:
                missing0 = int(m.get("proto_missing_0", 0))
                missing1 = int(m.get("proto_missing_1", 0))
                raw0     = m.get("proto_0_vec_json")
                raw1     = m.get("proto_1_vec_json")
                try:
                    if missing0 == 0 and raw0 is not None:
                        mu0 = np.array(
                            json.loads(raw0), dtype=np.float32)
                    if missing1 == 0 and raw1 is not None:
                        mu1 = np.array(
                            json.loads(raw1), dtype=np.float32)
                except Exception:
                    mu0 = mu1 = None
                if mu0 is None or mu1 is None:
                    invalid_proto.add(cid)

            mu0_list.append(mu0)
            mu1_list.append(mu1)

        dl2_arr = np.asarray(dl2_list, dtype=np.float32)
        wdata   = self._wdata(cid_list, n_by_cid)

        # ============================================================
        # ScoreRef
        # ============================================================
        base_ref             = np.zeros(len(cid_list), dtype=np.float32)
        score_ref_weighted   = np.zeros(len(cid_list), dtype=np.float32)
        score_ref_unweighted = np.zeros(len(cid_list), dtype=np.float32)
        cv_weighted = cv_unweighted = min_score_ref_cv = \
            score_ref_cv_final = 0.0
        cv_source = "NO_REF"
        cv_used   = 0.0
        use_weighted = False

        if self.mu_ref is not None:
            for i, cid in enumerate(cid_list):
                mu0, mu1 = mu0_list[i], mu1_list[i]
                if mu0 is None or mu1 is None:
                    continue
                m0   = self._cos(mu0, self.mu_ref[0]) - \
                       self._cos(mu0, self.mu_ref[1])
                m1   = self._cos(mu1, self.mu_ref[1]) - \
                       self._cos(mu1, self.mu_ref[0])
                base = float(min(m0, m1))
                base_ref[i]             = base
                score_ref_unweighted[i] = base
                score_ref_weighted[i]   = base * float(wdata[i])
            nonneg_w = score_ref_weighted   >= 0.0
            nonneg_u = score_ref_unweighted >= 0.0
            cv_weighted   = float(self._cv_score(
                score_ref_weighted[nonneg_w])) \
                if np.sum(nonneg_w) >= 2 else 0.0
            cv_unweighted = float(self._cv_score(
                score_ref_unweighted[nonneg_u])) \
                if np.sum(nonneg_u) >= 2 else 0.0
            if cv_weighted <= cv_unweighted:
                use_weighted    = True
                score_ref_split = score_ref_weighted.copy()
                score_ref       = score_ref_weighted.copy()
                cv_source = "WEIGHTED"
                cv_used   = cv_weighted
            else:
                use_weighted    = False
                score_ref_split = score_ref_unweighted.copy()
                score_ref       = score_ref_unweighted.copy()
                cv_source = "UNWEIGHTED"
                cv_used   = cv_unweighted
            min_score_ref_cv   = float(
                min(cv_weighted, cv_unweighted))
            score_ref_cv_final = float(min_score_ref_cv)
        else:
            score_ref = score_ref_split = base_ref = np.zeros(
                len(cid_list), dtype=np.float32)

        if self.mu_ref is None:
            (highref, lowref, thr, s_range, s_med,
             cv_ref, score_cv_thr_round, ref_mode) = (
                cid_list[:], [], 0.0, 0.0, 0.0, 0.0,
                self.score_cv_floor, "NO_REF")
            cv_base = 0.0
        else:
            (highref, lowref, thr, s_range, s_med,
             cv_ref, score_cv_thr_round, ref_mode) = \
                self._dynamic_ref_split(
                    cid_list, score_ref_split, score_ref_cv_final)
            base_nonneg = np.asarray(
                [base_ref[i] for i in range(len(cid_list))
                 if base_ref[i] >= 0.0],
                dtype=np.float32)
            base_nonneg = base_nonneg[np.isfinite(base_nonneg)]
            cv_base = float(self._cv_score(base_nonneg)) \
                if base_nonneg.size >= 2 else 0.0

        self.score_cv_hist.append(float(cv_ref))
        for i, cid in enumerate(cid_list):
            if mu0_list[i] is not None and mu1_list[i] is not None:
                self._client_scoreref_hist[cid].append(float(score_ref[i]))

        # ============================================================
        # Global prototypes
        # ============================================================
        proto_builders = [
            c for c in highref if c not in invalid_proto]
        if not proto_builders:
            proto_builders = [
                c for c in cid_list
                if c not in invalid_proto and c not in lowref]
        if not proto_builders:
            proto_builders = [
                c for c in cid_list if c not in invalid_proto]

        g0, g1 = self._build_global_proto_geomed(
            cid_list, mu0_list, mu1_list, n_by_cid, proto_builders)
        if (g0 is None or g1 is None) and \
                self._prev_global_proto is not None:
            g0, g1 = self._prev_global_proto
        if g0 is not None and g1 is not None:
            self._prev_global_proto = (g0.copy(), g1.copy())

        # ============================================================
        # Sep / Swap
        # ============================================================
        m0_g   = np.zeros(len(cid_list), dtype=np.float32)
        m1_g   = np.zeros(len(cid_list), dtype=np.float32)
        sep    = np.zeros(len(cid_list), dtype=np.float32)
        swap_v = np.zeros(len(cid_list), dtype=np.float32)

        if g0 is not None and g1 is not None:
            for i, cid in enumerate(cid_list):
                mu0, mu1 = mu0_list[i], mu1_list[i]
                if mu0 is None or mu1 is None:
                    continue
                c00 = self._cos(mu0, g0)
                c01 = self._cos(mu0, g1)
                c11 = self._cos(mu1, g1)
                c10 = self._cos(mu1, g0)
                m0_g[i]   = c00 - c01
                m1_g[i]   = c11 - c10
                if m0_g[i] < 0 or m1_g[i] < 0:
                    sep[i] = 0.0
                else:
                    sep[i] = max(0.0, 0.5 * (m0_g[i] + m1_g[i]))
                swap_v[i] = float(
                    max(0.0, c01 - c00) + max(0.0, c10 - c11))

        # ============================================================
        # Δsep CV gate
        # ============================================================
        idx           = {cid: i for i, cid in enumerate(cid_list)}
        honest_for_cv = [c for c in highref if c in idx]
        if len(honest_for_cv) < 2:
            honest_for_cv = [c for c in cid_list if c in idx]

        sep_h  = np.asarray(
            [sep[idx[c]] for c in honest_for_cv],
            dtype=np.float32)
        cv_now = mad_cv(sep_h, eps=self.cv_eps)
        cv_thr = winsorized_max_thr(
            self.cv_hist,
            floor=self.cv_floor,
            alpha=self.cv_win_alpha)
        sep_ok = bool(cv_now <= cv_thr)

        stable = False
        if self.use_cv_gate and (len(lowref) == 0):
            stable = bool(cv_now <= cv_thr)

        self.cv_hist.append(float(cv_now))

        # ============================================================
        # Trust scores
        # ============================================================
        if stable:
            sem_trust = np.clip(
                sigmoid(3.0 * sep), 0.0, 1.0
            ).astype(np.float32)
        else:
            q_trust   = sep - self.trust_lambda_swap * swap_v
            sem_trust = np.clip(q_trust, 0.0, 1.0).astype(np.float32)
        trust_round: Dict[int, float] = {
            cid_list[i]: float(np.clip(
                float(sem_trust[i]),
                self.min_trust, self.max_trust))
            for i in range(len(cid_list))
        }
        # ============================================================
        # Dual clustering setup
        # ============================================================
        suspicious_trust: List[int] = []
        suspicious_dsep:  List[int] = []
        trust_flag: Dict[int, int]  = {cid: 0 for cid in cid_list}

        kT = kD = 1
        km_dbg_T = km_dbg_D    = None
        low_center_T_print      = None
        labelsT = labelsD       = None
        order_mapT = order_mapD = None

        valid_for_km = [
            c for c in cid_list if c not in invalid_proto]
        n_valid = len(valid_for_km)

        def _finite_filter(valid_ids, vals):
            vals = np.asarray(vals, dtype=np.float32).reshape(-1)
            mask = np.isfinite(vals)
            return ([valid_ids[i] for i in range(len(valid_ids))
                     if bool(mask[i])], vals[mask])

        def _cluster(vals, apply_outlier):
            if self.use_outlier_bic and apply_outlier:
                return kmeans_bic_outlier_directional_1d(
                    vals,
                    seed=self.kmeans_random_state,
                    n_init=self.kmeans_n_init,
                    probe_min_cluster_size=
                        self.kmeans_probe_min_cluster_size,
                    core_min_cluster_size=
                        self.kmeans_core_min_cluster_size,
                    keep_low_singleton_if_core_k1=
                        self.keep_low_singleton_if_core_k1,
                )
            return kmeans_bic_hard_split_1d_guarded(
                vals,
                seed=self.kmeans_random_state,
                n_init=self.kmeans_n_init,
                min_cluster_size=self.kmeans_core_min_cluster_size,
            )

        def _low_side_ids(valid_ids, labels, order_map):
            return [
                valid_ids[i] for i in range(len(valid_ids))
                if int(order_map[int(labels[i])]) == 0]
        

        # ============================================================
        # Run dual clustering
        # ============================================================
        if self.use_trust_for_flagging and \
                (not stable) and n_valid >= 2:

            # Clustering T — Trust scores
            Tvals = np.asarray(
                [trust_round[c] for c in valid_for_km],
                dtype=np.float32)
            validT_ids, Tvals2 = _finite_filter(valid_for_km, Tvals)
            if len(validT_ids) >= 2:
                kT, labelsT, centers_sortedT, order_mapT, \
                    km_dbg_T = _cluster(
                        Tvals2, self.outlier_apply_to_trust)
                if centers_sortedT is not None and \
                        len(centers_sortedT):
                    v0 = float(centers_sortedT[0])
                    low_center_T_print = \
                        v0 if np.isfinite(v0) else None
                if kT == 2:
                    suspicious_trust = _low_side_ids(
                        validT_ids, labelsT, order_mapT)

            # Clustering D — Δsep scores
            Dvals = np.asarray(
                [float(sep[idx[c]]) for c in valid_for_km],
                dtype=np.float32)
            validD_ids, Dvals2 = _finite_filter(valid_for_km, Dvals)
            if len(validD_ids) >= 2:
                kD, labelsD, centers_sortedD, order_mapD, \
                    km_dbg_D = _cluster(
                        Dvals2, self.outlier_apply_to_dsep)
                if kD == 2:
                    suspicious_dsep = _low_side_ids(
                        validD_ids, labelsD, order_mapD)

            # Intersection
            suspicious_final = sorted(
                set(suspicious_trust).intersection(
                    set(suspicious_dsep)))
            print(f"\n[Clustering] kT={kT} kD={kD} | "
                  f"suspicious_T={sorted(suspicious_trust)} | "
                  f"suspicious_D={sorted(suspicious_dsep)} | "
                  f"intersection={suspicious_final}")

            trust_flag = {
                cid: (1 if cid in suspicious_final else 0)
                for cid in cid_list}

        else:
            suspicious_final = []
            print(f"\n[Clustering] Skipped — "
                  f"stable={stable} n_valid={n_valid}")

        malicious = sorted(set(suspicious_final))
        honest    = sorted(set(cid_list) - set(malicious))

        # ============================================================
        # Adaptive ScoreRef FlagGate
        # ============================================================

        print(f"\n[FlagGate] per-client score_ref history check")
        
        # Step 2 — Apply gate
        # If a clustering-flagged client passes the gate (swap=0 AND
        # ScoreRef >= gate), it is moved back to honest.
        # This means: included in aggregation, status OK, flag count
        # unchanged, and ScoreRef added to honest history.
        suppressed_by_gate: Set[int] = set()
        for cid in list(malicious):
            c_idx          = idx[cid]
            has_swap       = bool(swap_v[c_idx] > 0)
            prev_rounds    = self._client_scoreref_hist.get(cid, [])[:-1]
            was_neg_before = any(s < 0.0 for s in prev_rounds)
        
            # ── NEW ──────────────────────────────────────────────────
            is_round1 = len(prev_rounds) == 0   # no history = first time seen
        
            if is_round1:
                confirmed = has_swap or bool(score_ref[c_idx] < 0.0)
            else:
                confirmed = has_swap or was_neg_before
            # ─────────────────────────────────────────────────────────
        
            if confirmed:
                print(
                    f"  [FlagGate] CID {cid} CONFIRMED — "
                    f"swap={swap_v[c_idx]:.4f} "
                    f"ScoreR={score_ref[c_idx]:.4f} "
                    f"was_neg_before={was_neg_before}"
                )
            else:
                suppressed_by_gate.add(cid)
                print(
                    f"  [FlagGate] CID {cid} SUPPRESSED → honest — "
                    f"swap={swap_v[c_idx]:.4f} "
                    f"ScoreR={score_ref[c_idx]:.4f} "
                    f"was_neg_before={was_neg_before}"
                )

        # Step 3 — Update malicious and honest after gate
        malicious = sorted(set(malicious) - suppressed_by_gate)
        honest    = sorted(set(cid_list) - set(malicious))

        if suppressed_by_gate:
            print(f"[FlagGate] After gate — "
                  f"malicious={malicious} "
                  f"honest={honest} "
                  f"suppressed={sorted(suppressed_by_gate)}")

        # ============================================================
        # Strikes / banning
        # ============================================================
        newly_banned: List[int] = []
        for cid in cid_list:
            if cid in invalid_proto:
                self.missing_proto_count[cid] += 1
            else:
                self.missing_proto_count[cid] = 0

            if cid in malicious:
                self.client_flag_count[cid] += 1

            if self.missing_proto_count[cid] >= \
                    self.missing_proto_strike_after:
                self.banned_clients.add(cid)
                newly_banned.append(cid)
            elif self.client_flag_count[cid] >= \
                    self.ban_after_flags:
                self.banned_clients.add(cid)
                newly_banned.append(cid)

        if newly_banned:
            print(f"\n[Banning] Newly banned this round: "
                  f"{sorted(newly_banned)}")

        # ============================================================
        # Aggregation pool
        # ============================================================
        good_cids = [
            c for c in honest
            if c not in self.banned_clients
            and n_by_cid.get(c, 0) > 0]
        if len(good_cids) == 0:
            return None, {
                "banned_count":    int(len(self.banned_clients)),
                "malicious_count": int(len(malicious))
            }

        print(f"\n[DEBUG AGG Round {rnd}] good_cids={good_cids}")
        for c in good_cids:
            p = parameters_to_ndarrays(fit_by_cid[c].parameters)
            print(f"  Client {c} params[0] "
                  f"mean={float(p[0].mean()):.6f} "
                  f"std={float(p[0].std()):.6f}")

        agg_pool = [
            (parameters_to_ndarrays(fit_by_cid[c].parameters),
             int(n_by_cid.get(c, 0)))
            for c in good_cids
        ]
        total    = sum(n for _, n in agg_pool) or 1
        template = agg_pool[0][0]
        agg_ndarrays = [np.zeros_like(w) for w in template]
        for w, n in agg_pool:
            a = n / total
            for k in range(len(w)):
                agg_ndarrays[k] += w[k] * a

        aggregated_parameters = ndarrays_to_parameters(agg_ndarrays)
        # ============================================================
        # Save to FEDSEM_HIST
        # ============================================================
        FEDSEM_GLOBAL_FINAL["params"] = parameters_to_ndarrays(
            aggregated_parameters)

        for cp, fr in results:
            cid_str = str(self._cid_from(cp, fr))
            _init_client_history(cid_str)
            m = fr.metrics or {}
            h = FEDSEM_HIST["clients"][cid_str]
            h["round"].append(int(rnd))
            h["train_loss"].append(
                _safe_float(m.get("train_loss", 0.0)))
            h["train_acc"].append(
                _safe_float(m.get("train_acc", 0.0)))
            h["val_loss"].append(
                _safe_float(m.get("val_loss", 0.0)))
            h["val_acc"].append(
                _safe_float(m.get("val_acc", 0.0)))
            h["avg_local_loss"].append(
                _safe_float(m.get("avg_local_loss", 0.0)))
            h["best_val_loss"].append(
                _safe_float(m.get("best_val_loss", np.nan), np.nan))
            h["stopped_epoch"].append(
                _safe_int(m.get("stopped_epoch", -1), -1))
            h["early_stopped"].append(
                _safe_int(m.get("early_stopped", 0)))
            h["train_loss_curve_json"].append(
                _safe_json_or_none(
                    m.get("train_loss_curve_json")))
            h["val_loss_curve_json"].append(
                _safe_json_or_none(m.get("val_loss_curve_json")))
            h["n_train"].append(
                _safe_int(m.get("n_train", fr.num_examples)))
            h["n_val"].append(_safe_int(m.get("n_val", 0)))
            h["n_test"].append(_safe_int(m.get("n_test", 0)))
            h["has_protos"].append(
                _safe_int(m.get("has_protos", 0)))
            h["proto_sep_score"].append(
                _safe_float(m.get("proto_sep_score", 0.0)))
            h["proto_missing_0"].append(
                _safe_int(m.get("proto_missing_0", 0)))
            h["proto_missing_1"].append(
                _safe_int(m.get("proto_missing_1", 0)))
            h["proto_n0"].append(
                _safe_int(m.get("proto_n0", 0)))
            h["proto_n1"].append(
                _safe_int(m.get("proto_n1", 0)))
            h["proto_0_norm"].append(
                _safe_float(m.get("proto_0_norm", 0.0)))
            h["proto_1_norm"].append(
                _safe_float(m.get("proto_1_norm", 0.0)))
            h["proto_sep_cos"].append(
                _safe_float(m.get("proto_sep_cos", 0.0)))
            h["proto_0_vec_json"].append(
                m.get("proto_0_vec_json", None))
            h["proto_1_vec_json"].append(
                m.get("proto_1_vec_json", None))
            h["update_l2"].append(
                _safe_float(m.get("update_l2", 0.0)))
            h["banned_client"].append(
                1 if self._cid_from(cp, fr)
                in self.banned_clients else 0)

        FEDSEM_HIST["server"]["round"].append(int(rnd))
        FEDSEM_HIST["server"]["active_clients"].append(
            int(len(results)))
        FEDSEM_HIST["server"]["banned_clients"].append(
            int(len(self.banned_clients)))
        FEDSEM_HIST["server"]["failures"].append(
            int(len(failures)))
        FEDSEM_HIST["server"]["avg_local_loss_w"].append(
            float(np.mean([
                _safe_float(
                    (fr.metrics or {}).get("avg_local_loss", 0.0))
                for _, fr in results
            ]))
        )

        # ============================================================
        # Full diagnostic print
        # ============================================================
        trust_flag_mode = ("NOFLAG_STABLE" if stable
                           else f"KMEANS_BIC(kT={kT},kD={kD})")
        if self.use_outlier_bic and (not stable):
            trust_flag_mode += (
                "+DIR_OUTLIER_BIC+MAD_VAL"
                "+SINGLETON_LOWREF+ADAPTIVE_FLAGGATE")

        print(f"\n==================== [Server][Round {rnd}] "
              f"SUMMARY ====================")
        print(f"\n[Round {rnd}] Weight check:")
        for cid in cid_list:
            params     = parameters_to_ndarrays(
                fit_by_cid[cid].parameters)
            first_mean = float(params[0].mean())
            print(f"  Client {cid} first_mean={first_mean:.6f}")
        print(f"Proto builders       : {sorted(proto_builders)}")
        print(f"HighRef              : {sorted(highref)}")
        print(f"LowRef               : {sorted(lowref)}")
        print(f"Trust suspicious(T)  : {sorted(suspicious_trust)}")
        print(f"Δsep  suspicious(D)  : {sorted(suspicious_dsep)}")
        print(f"Final malicious      : {sorted(malicious)}")
        print(f"Honest clients       : {sorted(honest)}")
        print(f"Newly banned         : {sorted(newly_banned)}")
        print(f"Banned pool          : {sorted(self.banned_clients)}")

        print(
            f"\n[SCORE REF]\n"
            f"  mode               : {ref_mode}\n"
            f"  cv_ref(split)      : {cv_ref*100:.2f}%\n"
            f"  cv_base(dbg)       : {cv_base*100:.2f}%\n"
            f"  cv_weighted        : {cv_weighted*100:.2f}%\n"
            f"  cv_unweighted      : {cv_unweighted*100:.2f}%\n"
            f"  min_score_ref_cv   : {min_score_ref_cv*100:.2f}%\n"
            f"  using              : {cv_source} "
            f"(cv={cv_used*100:.2f}%)\n"
            f"  thrWin             : {score_cv_thr_round*100:.2f}%  "
            f"(floor={self.score_cv_floor*100:.2f}%, "
            f"α={self.score_cv_win_alpha})\n"
            f"  thr(med-alpha*MAD) : {thr:.4f}\n"
            f"  range/median       : {s_range:.4f} / {s_med:.4f}"
        )

        print(
            f"\n[Δsep CV]\n"
            f"  honest_for_cv      : {sorted(honest_for_cv)}\n"
            f"  cv_now (MAD-CV)    : {cv_now*100:.2f}%\n"
            f"  thrWin             : {cv_thr*100:.2f}%\n"
            f"  stable             : {stable}\n"
            f"  sep_ok(main)       : {sep_ok}\n"
            f"  singleton_cancel   : lowref_intersection "
            f"(cv_brule removed)"
        )

        print(
            f"\n[TRUST]\n"
            f"  trust_calc_mode    : SEP_MINUS_SWAP\n"
            f"  trust_flag_mode    : {trust_flag_mode}\n"
            f"  trust_lambda_swap  : {self.trust_lambda_swap}\n"
            f"  low-center(T)      : "
            f"{'NA' if low_center_T_print is None else f'{low_center_T_print:.4f}'}\n"
            f"  n_valid_for_km     : {n_valid}"
        )

        def _dbg_outlier_cid(valid_ids, dbg):
            if not isinstance(dbg, dict):
                return None
            out_idx = dbg.get("outlier_idx") or []
            if isinstance(out_idx, list) and len(out_idx) == 1:
                j = int(out_idx[0])
                if 0 <= j < len(valid_ids):
                    return int(valid_ids[j])
            return None

        if (not stable) and isinstance(km_dbg_T, dict):
            ocidT = _dbg_outlier_cid(valid_for_km, km_dbg_T)
            print(
                f"  [KM TRUST DBG] "
                f"out_idx={km_dbg_T.get('outlier_idx')} "
                f"out_cid={ocidT} "
                f"out_side={km_dbg_T.get('outlier_side')}\n"
                f"    probe: k_raw="
                f"{((km_dbg_T.get('probe_dbg') or {}).get('choose_k_raw'))} "
                f"k_final="
                f"{((km_dbg_T.get('probe_dbg') or {}).get('choose_k_final'))} "
                f"bic1="
                f"{((km_dbg_T.get('probe_dbg') or {}).get('bic1')):.3f} "
                f"bic2="
                f"{((km_dbg_T.get('probe_dbg') or {}).get('bic2')):.3f} "
                f"c_low/high="
                f"{((km_dbg_T.get('probe_dbg') or {}).get('c_low'))}"
                f"/{((km_dbg_T.get('probe_dbg') or {}).get('c_high'))}\n"
                f"    core : k_raw="
                f"{((km_dbg_T.get('core_dbg') or {}).get('choose_k_raw'))} "
                f"k_final="
                f"{((km_dbg_T.get('core_dbg') or {}).get('choose_k_final'))} "
                f"forced_k2={km_dbg_T.get('forced_k2_reason')} "
                f"centers={km_dbg_T.get('centers_final')}"
            ) 

        if (not stable) and isinstance(km_dbg_D, dict):
            ocidD = _dbg_outlier_cid(valid_for_km, km_dbg_D)
            print(
                f"  [KM ΔSEP DBG] "
                f"out_idx={km_dbg_D.get('outlier_idx')} "
                f"out_cid={ocidD} "
                f"out_side={km_dbg_D.get('outlier_side')}\n"
                f"    probe: k_raw="
                f"{((km_dbg_D.get('probe_dbg') or {}).get('choose_k_raw'))} "
                f"k_final="
                f"{((km_dbg_D.get('probe_dbg') or {}).get('choose_k_final'))} "
                f"bic1="
                f"{((km_dbg_D.get('probe_dbg') or {}).get('bic1')):.3f} "
                f"bic2="
                f"{((km_dbg_D.get('probe_dbg') or {}).get('bic2')):.3f} "
                f"c_low/high="
                f"{((km_dbg_D.get('probe_dbg') or {}).get('c_low'))}"
                f"/{((km_dbg_D.get('probe_dbg') or {}).get('c_high'))}\n"
                f"    core : k_raw="
                f"{((km_dbg_D.get('core_dbg') or {}).get('choose_k_raw'))} "
                f"k_final="
                f"{((km_dbg_D.get('core_dbg') or {}).get('choose_k_final'))} "
                f"forced_k2={km_dbg_D.get('forced_k2_reason')} "
                f"centers={km_dbg_D.get('centers_final')}"
            )

        print(f"\n[Server][Round {rnd}] === {self.title} ===")
        hdr = (
            f"{'CID':>3} | {'N':>6} | {'wD':>5} | {'ScoreR':>7} | "
            f"{'m0_g':>6} {'m1_g':>6} | {'Swap':>6} {'Trust':>7} | "
            f"{'TFlg':>4} {'Flag#':>5} | {'dL2':>10} | "
            f"{'Agg':>8} | {'Status':>8}"
        )
        print(hdr)
        print("-" * len(hdr))

        good_set = set(good_cids)
        for i, cid in enumerate(cid_list):
            status    = (
                "BANNED"  if cid in self.banned_clients else
                "FLAGGED" if cid in malicious else "OK")
            agg_state = (
                "USED"   if cid in good_set else
                "BANNED" if cid in self.banned_clients
                else "EXCLUDED")
            print(
                f"{cid:>3} | {n_by_cid.get(cid,0):6d} | "
                f"{wdata[i]:5.2f} | "
                f"{score_ref[i]:7.3f} | "
                f"{m0_g[i]:6.3f} {m1_g[i]:6.3f} | "
                f"{swap_v[i]:6.3f} {trust_round[cid]:7.4f} | "
                f"{trust_flag.get(cid,0):4d} "
                f"{int(self.client_flag_count[cid]):5d} | "
                f"{dl2_arr[i]:10.3e} | "
                f"{agg_state:>8} | {status:>8}"
            )

        def _f(x, nd=4):
            return ("   nan" if not np.isfinite(x)
                    else f"{x:7.{nd}f}")

        print(f"\n[Server][Round {rnd}] === "
              f"Client Learning Performance ===")
        hdr2 = (
            f"{'CID':>3} | {'TrLoss':>7} {'TrAcc':>6} | "
            f"{'VaLoss':>7} {'VaAcc':>6} | "
            f"{'TeLoss':>7} {'TeAcc':>6} | "
            f"{'ES':>2} {'StopEp':>6} {'BestVL':>7} | "
            f"{'N':>7} | {'n0':>5} {'n1':>5} | {'Role':>6}"
        )
        print(hdr2)
        print("-" * len(hdr2))

        for cp, fr in results:
            cid = self._cid_from(cp, fr)
            m   = fr.metrics or {}
            p   = perf_by_cid.get(cid, {})
            trL = float(p.get("train_loss", np.nan))
            trA = float(p.get("train_acc",  np.nan))
            vaL = float(p.get("val_loss",   np.nan))
            vaA = float(p.get("val_acc",    np.nan))
            teL = float(p.get("test_loss",  np.nan))
            teA = float(p.get("test_acc",   np.nan))
            es       = _safe_int(m.get("early_stopped", 0))
            stop_ep  = _safe_int(m.get("stopped_epoch", -1), -1)
            best_vl  = _safe_float(
                m.get("best_val_loss", np.nan), np.nan)
            best_vl_str = (f"{best_vl:7.3f}"
                           if np.isfinite(best_vl) else "    nan")
            role = (
                "BAN" if cid in self.banned_clients else
                "MAL" if cid in malicious else "HON")
            print(
                f"{cid:>3} | {_f(trL)} "
                f"{('  nan' if not np.isfinite(trA) else f'{trA:6.3f}')} | "
                f"{_f(vaL)} "
                f"{('  nan' if not np.isfinite(vaA) else f'{vaA:6.3f}')} | "
                f"{_f(teL)} "
                f"{('  nan' if not np.isfinite(teA) else f'{teA:6.3f}')} | "
                f"{es:2d} {stop_ep:6d} {best_vl_str} | "
                f"{n_by_cid.get(cid,0):7d} | "
                f"{n0_by_cid.get(cid,0):5d} "
                f"{n1_by_cid.get(cid,0):5d} | {role:>6}"
            )

        print(f"\n[Server][Round {rnd}] === "
              f"Per-Client Loss Curves ===")
        for cp, fr in results:
            cid = self._cid_from(cp, fr)
            m   = fr.metrics or {}
            try:
                tr_curve = json.loads(
                    m.get("train_loss_curve_json") or "[]")
                va_curve = json.loads(
                    m.get("val_loss_curve_json") or "[]")
            except Exception:
                tr_curve = []
                va_curve = []
            role = (
                "BAN" if cid in self.banned_clients else
                "MAL" if cid in malicious else "HON")
            print(f"  Client {cid} [{role}]")
            for ep_idx, (tl, vl) in enumerate(
                    zip(tr_curve, va_curve)):
                print(f"    Epoch {ep_idx+1}/{len(tr_curve)} "
                      f"— Train Loss: {tl:.4f} "
                      f"| Val Loss: {vl:.4f}")
            if not tr_curve:
                print(f"    (no curve data)")

        metrics_agg: Dict[str, Scalar] = {
            "banned_count":
                int(len(self.banned_clients)),
            "malicious_count":
                int(len(malicious)),
            "stable":
                int(stable),
            "sep_ok_main":
                int(sep_ok),
            "score_cv_split":
                float(cv_ref),
            "score_cv_base_dbg":
                float(cv_base),
            "cv_weighted":
                float(cv_weighted),
            "cv_unweighted":
                float(cv_unweighted),
            "min_score_ref_cv":
                float(min_score_ref_cv),
            "score_ref_cv_final":
                float(score_ref_cv_final),
            "use_weighted_for_split":
                int(use_weighted),
            "sep_cv":
                float(cv_now),
            "sep_cv_thr_win":
                float(cv_thr),
            "k_trust":
                int(kT),
            "k_dsep":
                int(kD),
            "trust_lambda_swap":
                float(self.trust_lambda_swap),
            "use_outlier_bic":
                int(self.use_outlier_bic),
            "keep_low_singleton_if_core_k1":
                int(self.keep_low_singleton_if_core_k1),
            "singleton_cancel_rule":
                "lowref_intersection",
        }

        return aggregated_parameters, metrics_agg
```

```python
import json
import inspect
import numpy as np
import torch
import torch.nn.functional as F
from torch.utils.data import DataLoader
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, cohen_kappa_score, roc_auc_score,
    confusion_matrix,
)
import flwr as fl
from flwr.common import parameters_to_ndarrays, ndarrays_to_parameters
from sklearn.model_selection import train_test_split


TOTAL_CPUS = 1


# -------------------- JSON helper --------------------
def json_default(o):
    if isinstance(o, (np.integer,)):  return int(o)
    if isinstance(o, (np.floating,)): return float(o)
    if isinstance(o, (np.ndarray,)):  return o.tolist()
    return str(o)

def _fmt(v, nd=4):
    if v is None: return "N/A"
    try:    return f"{float(v):.{nd}f}"
    except: return str(v)


# ============================================================
# Worker-safe dataset access
# ============================================================
_PRINT_ONCE = set()

def get_df_local(cid_int: int):
    cds, _, _ = get_client_datasets_cached(verbose=False)
    if cid_int not in cds:
        raise KeyError(
            f"cid {cid_int} not found in cds keys="
            f"{sorted(list(cds.keys()))}")
    return cds[cid_int]

def get_df_global_test_worker():
    _, _, df_test = get_client_datasets_cached(verbose=False)
    return df_test


# ============================================================
# Global val + test loaders (server-side centralized evaluation)
# ============================================================

def build_global_eval_loaders(batch_size: int = 32):
    _, df_global_val, df_global_test = get_client_datasets_cached(
        verbose=False)

    global_val_loader = DataLoader(
        BanglaSentimentDataset(df_global_val,  "review_clean", "y"),
        batch_size = batch_size,
        shuffle    = False,
    )
    global_test_loader = DataLoader(
        BanglaSentimentDataset(df_global_test, "review_clean", "y"),
        batch_size = batch_size,
        shuffle    = False,
    )

    print(f"[Server] Global val  : {len(df_global_val)} samples")
    print(f"[Server] Global test : {len(df_global_test)} samples")

    return global_val_loader, global_test_loader


# ============================================================
# Client function
# ============================================================

def client_fn(context, ) -> fl.client.Client:
    from flwr.common import Context
    pid = context.node_config.get("partition-id", None)
    if pid is None:
        raise KeyError(
            "context.node_config missing 'partition-id'")
    cid_int  = int(pid)
    df_local = get_df_local(cid_int)

    if cid_int not in _PRINT_ONCE:
        _PRINT_ONCE.add(cid_int)
        try:    y_counts = df_local["y"].value_counts().to_dict()
        except: y_counts = {}
        print(f"[Client {cid_int}] Init N={len(df_local)} "
              f"| y_counts={y_counts}")

    # ── All data used for training ───────────────────────────
    # No val split — early stopping removed
    # Server handles all evaluation centrally via evaluate_fn
    df_train = df_local.reset_index(drop=True)
    # ────────────────────────────────────────────────────────

    model   = init_banglabert_classifier(num_labels=NUM_LABELS)
    encoder = init_text_encoder_from_model(model)

    return FedSemClient(
        cid        = cid_int,
        model      = model,
        encoder    = encoder,
        train_df   = df_train,
        val_df     = None,      # all data used for training
        test_df    = None,      # evaluate() disabled
        num_labels = NUM_LABELS,
    ).to_client()


# ============================================================
# Helper: apply final aggregated weights to a model
# ============================================================

def set_model_weights(model, ndarrays):
    with torch.no_grad():
        dummy_encoder = init_text_encoder_from_model(model)
        state_dict    = get_trainable_state_dict(
            model, fusion_encoder=dummy_encoder)
        param_keys    = list(state_dict.keys())
        new_state     = {
            k: torch.tensor(v)
            for k, v in zip(param_keys, ndarrays)}
        apply_trainable_state_dict(
            model, new_state, fusion_encoder=dummy_encoder)


# ============================================================
# Helper: full evaluation metrics
# ============================================================

@torch.no_grad()
def evaluate_all_metrics(model, loader, num_labels, device):
    model.eval()
    all_preds  = []
    all_labels = []
    all_probs  = []
    _use_amp   = device.type == "cuda"

    for batch in loader:
        input_ids      = batch["input_ids"].to(device)
        attention_mask = batch["attention_mask"].to(device)
        labels         = batch["labels"].to(device)

        fwd = {
            "input_ids":      input_ids,
            "attention_mask": attention_mask,
            "return_dict":    True,
        }
        if "token_type_ids" in batch:
            fwd["token_type_ids"] = batch["token_type_ids"].to(device)

        with torch.cuda.amp.autocast(enabled=_use_amp):
            out = model(**fwd)

        probs = F.softmax(out.logits, dim=-1)
        preds = torch.argmax(probs, dim=-1)

        all_preds.append(preds.cpu().numpy())
        all_labels.append(labels.cpu().numpy())
        all_probs.append(probs.cpu().numpy())

    all_preds  = np.concatenate(all_preds)
    all_labels = np.concatenate(all_labels)
    all_probs  = np.concatenate(all_probs)

    acc       = float(accuracy_score(all_labels, all_preds))
    precision = float(precision_score(
        all_labels, all_preds, average="macro", zero_division=0))
    recall    = float(recall_score(
        all_labels, all_preds, average="macro", zero_division=0))
    f1        = float(f1_score(
        all_labels, all_preds, average="macro", zero_division=0))
    kappa     = float(cohen_kappa_score(all_labels, all_preds))
    cm        = confusion_matrix(all_labels, all_preds).tolist()

    roc_auc = 0.0
    try:
        if num_labels == 2:
            roc_auc = float(
                roc_auc_score(all_labels, all_probs[:, 1]))
        else:
            roc_auc = float(roc_auc_score(
                all_labels, all_probs, multi_class="ovr"))
    except Exception:
        pass

    return {
        "acc": acc, "precision": precision, "recall": recall,
        "f1": f1, "kappa": kappa, "roc_auc": roc_auc,
        "confusion_matrix": cm,
    }


# ============================================================
# Post-simulation summary printer
# ============================================================

def print_post_simulation_summary():
    print("\n" + "=" * 70)
    print("POST-SIMULATION SUMMARY — ALL ROUNDS")
    print("=" * 70)

    print("\n[Server Round History]")
    hdr = (
        f"{'Round':>5} | {'ActiveClients':>13} | {'Banned':>6} | "
        f"{'AvgLoss':>7} | {'Failures':>8}"
    )
    print(hdr)
    print("-" * len(hdr))
    srv = FEDSEM_HIST["server"]
    for i, rnd in enumerate(srv["round"]):
        print(
            f"{rnd:>5} | "
            f"{srv['active_clients'][i]:>13} | "
            f"{srv['banned_clients'][i]:>6} | "
            f"{srv['avg_local_loss_w'][i]:>7.4f} | "
            f"{srv['failures'][i]:>8}"
        )

    print("\n[Per-Client Round Summary]")
    for cid_str, h in sorted(
            FEDSEM_HIST["clients"].items(),
            key=lambda x: int(x[0])):
        print(f"\n  Client {cid_str}:")
        hdr2 = (
            f"    {'Rnd':>3} | {'TrLoss':>7} {'TrAcc':>6} | "
            f"{'VaLoss':>7} {'VaAcc':>6} | "
            f"{'ES':>2} {'StopEp':>6} {'BestVL':>7} | {'Banned':>6}"
        )
        print(hdr2)
        print("    " + "-" * (len(hdr2) - 4))
        for i, rnd in enumerate(h["round"]):
            trl     = h["train_loss"][i]     if i < len(h["train_loss"])     else float("nan")
            tra     = h["train_acc"][i]      if i < len(h["train_acc"])      else float("nan")
            val     = h["val_loss"][i]       if i < len(h["val_loss"])       else float("nan")
            vaa     = h["val_acc"][i]        if i < len(h["val_acc"])        else float("nan")
            es      = h["early_stopped"][i]  if i < len(h["early_stopped"])  else 0
            stop_ep = h["stopped_epoch"][i]  if i < len(h["stopped_epoch"])  else -1
            best_vl = h["best_val_loss"][i]  if i < len(h["best_val_loss"])  else float("nan")
            banned  = h["banned_client"][i]  if i < len(h["banned_client"])  else 0
            best_vl_str = f"{best_vl:7.3f}" if np.isfinite(best_vl) else "    nan"
            trl_str = f"{trl:7.4f}" if np.isfinite(trl) else "    nan"
            tra_str = f"{tra:6.3f}" if np.isfinite(tra) else "   nan"
            val_str = f"{val:7.4f}" if np.isfinite(val) else "    nan"
            vaa_str = f"{vaa:6.3f}" if np.isfinite(vaa) else "   nan"
            print(
                f"    {rnd:>3} | "
                f"{trl_str} {tra_str} | "
                f"{val_str} {vaa_str} | "
                f"{es:>2} {stop_ep:>6} {best_vl_str} | "
                f"{'YES' if banned else 'no':>6}"
            )

    # ── Global centralized evaluation history ────────────────
    if srv.get("global_val_loss"):
        print("\n[Global Centralized Evaluation per Round]")
        hdr3 = (
            f"{'Round':>5} | {'ValLoss':>8} {'ValAcc':>7} | "
            f"{'TestLoss':>9} {'TestAcc':>8}"
        )
        print(hdr3)
        print("-" * len(hdr3))
        for i, rnd in enumerate(srv["round"]):
            vl = srv["global_val_loss"][i]  if i < len(srv["global_val_loss"])  else float("nan")
            va = srv["global_val_acc"][i]   if i < len(srv["global_val_acc"])   else float("nan")
            tl = srv["global_test_loss"][i] if i < len(srv["global_test_loss"]) else float("nan")
            ta = srv["global_test_acc"][i]  if i < len(srv["global_test_acc"])  else float("nan")
            print(
                f"{rnd:>5} | "
                f"{vl:8.4f} {va:7.4f} | "
                f"{tl:9.4f} {ta:8.4f}"
            )

    print("\n" + "=" * 70)


# ============================================================
# Main simulation function
# ============================================================

def run_fedsem():

    # 0) Build datasets once on driver
    cds, df_val_global, df_test_global = get_client_datasets_cached(
        verbose=True)

    global_val_loader = DataLoader(
        BanglaSentimentDataset(
            df_val_global, "review_clean", "y"),
        batch_size = 32,
        shuffle    = False,
    )
    global_test_loader_eval = DataLoader(
        BanglaSentimentDataset(
            df_test_global, "review_clean", "y"),
        batch_size = 32,
        shuffle    = False,
    )
    print(f"[Server] Global val  : {len(df_val_global)} samples")
    print(f"[Server] Global test : {len(df_test_global)} samples")

    # 1) Trust server model — trained on df_trust
    df_trust_0        = df_trust[df_trust["y"] == 0]
    df_trust_1        = df_trust[df_trust["y"] == 1]
    min_count         = min(len(df_trust_0), len(df_trust_1))
    df_trust_balanced = pd.concat([
        df_trust_0.sample(min_count, random_state=SEED),
        df_trust_1.sample(min_count, random_state=SEED)
    ], ignore_index=True).reset_index(drop=True)
    print(f"[Trust] Balanced df_trust: "
          f"{df_trust_balanced['y'].value_counts().to_dict()}")

    trust_server_model = init_banglabert_classifier(
        num_labels=NUM_LABELS)
    trust_server_model.to(DEVICE)

    trust_train_loader = make_trust_loader(
        df_trust   = df_trust_balanced,
        tokenizer  = tokenizer,
        batch_size = 16,
        max_len    = ENC_MAX_LEN,
        shuffle    = True,
    )
    trust_val_loader = make_trust_loader(
        df_trust   = df_trust_balanced,
        tokenizer  = tokenizer,
        batch_size = 16,
        max_len    = ENC_MAX_LEN,
        shuffle    = False,
    )

    print("[Trust] Training server model on df_trust...")
    _, _, trust_history = train_one_fl_round(
        model              = trust_server_model,
        train_loader       = trust_train_loader,
        val_loader         = trust_val_loader,
        fusion_encoder     = None,
        device             = DEVICE,
        local_epochs       = 10,
        base_lr            = BASE_LR,
        adapter_lr         = ADAPTER_LR,
        weight_decay       = WEIGHT_DECAY,
        accumulation_steps = ACCUMULATION_STEPS,
        use_fp16           = USE_FP16,
        cid                = -1,
    )
    print(f"[Trust] ✅ Trained! "
          f"Best val loss: {trust_history['best_val_loss']:.4f}")
    trust_server_model.eval()

    # 2) Strategy signature validation
    sig  = str(inspect.signature(FedSemDefenceStrategy.__init__))
    must = [
        "use_outlier_bic",
        "keep_low_singleton_if_core_k1",
        "kmeans_probe_min_cluster_size",
        "kmeans_core_min_cluster_size",
        "trust_lambda_swap",
    ]
    assert all(k in sig for k in must), (
        f"OLD strategy loaded. Re-run strategy cell.\n"
        f"Current signature: {sig}"
    )
    assert "small_n_max" not in sig, (
        f"Old small_n logic still exists. Re-run strategy cell.\n"
        f"Current signature: {sig}"
    )

    # 3) Get initial parameters from trained server model
    dummy_encoder  = init_text_encoder_from_model(trust_server_model)
    trained_state  = get_trainable_state_dict(
        trust_server_model, fusion_encoder=dummy_encoder)
    initial_arrays = [v.cpu().numpy() for v in trained_state.values()]
    initial_params = ndarrays_to_parameters(initial_arrays)
    print(f"[Trust] ✅ Initial params ready — "
          f"{len(initial_arrays)} arrays")

    # ── param_keys for evaluate_fn ───────────────────────────
    with torch.no_grad():
        _init_state = get_trainable_state_dict(
            trust_server_model, fusion_encoder=dummy_encoder)
        param_keys = list(_init_state.keys())
    print(f"[Server] param_keys : {len(param_keys)} keys")

    global strategy_obj

    # 4) Strategy init
    strategy_obj = FedSemDefenceStrategy(
        fraction_fit                  = 1.0,
        fraction_evaluate             = 0.0,
        min_fit_clients               = 1,
        min_available_clients         = NUM_CLIENTS,
        ban_after_flags               = 5,
        ref_alpha_mad                 = 1.0,
        score_cv_eps                  = 1e-12,
        score_cv_floor                = 0.07,
        score_cv_win_alpha            = 2.5,
        geo_weight_by_examples        = True,
        missing_proto_strike_after    = 2,
        trust_lambda_swap             = 2,
        use_trust_for_flagging        = True,
        kmeans_random_state           = 0,
        kmeans_n_init                 = 50,
        kmeans_probe_min_cluster_size = 1,
        kmeans_core_min_cluster_size  = 1,
        use_outlier_bic               = True,
        keep_low_singleton_if_core_k1 = True,
        outlier_apply_to_trust        = True,
        outlier_apply_to_dsep         = True,
        use_cv_gate                   = True,
        cv_eps                        = 1e-12,
        cv_floor                      = 0.07,
        cv_win_alpha                  = 2.5,
        df_trust                      = df_trust_balanced,
        trust_tokenizer               = tokenizer,
        trust_server_model            = trust_server_model,
        trust_device                  = DEVICE,
        trust_batch_size              = 32,
        trust_max_len                 = 48,
        trust_mom_K                   = 5,
        trust_mom_seed                = 42,
        num_labels                    = NUM_LABELS,
        initial_parameters            = initial_params,
        evaluate_fn                   = get_evaluate_fn(
            model       = trust_server_model,
            encoder     = dummy_encoder,
            val_loader  = global_val_loader,
            test_loader = global_test_loader_eval,
            device      = DEVICE,
            param_keys  = param_keys,
        ),
        title=(
            "FedSemProDefence v3.8.0 "
            "(dual clustering + directional outlier BIC + "
            "MAD validation + adaptive ScoreRef FlagGate + "
            "singleton cancel via lowref intersection)"
        ),
    )

    strategy_obj.min_trust = 0.0
    strategy_obj.max_trust = 1.0

    # 5) Ray init args
    ray_init_args = {
        "include_dashboard":   False,
        "num_cpus":            int(TOTAL_CPUS),
        "num_gpus":            1,
        "log_to_driver":       False,
        "ignore_reinit_error": True,
    }

    # 6) Run simulation
    print(
        f"\n[run_fedsem] Starting FL simulation | "
        f"clients={NUM_CLIENTS} rounds={ROUNDS} device={DEVICE}"
    )

    fl.simulation.start_simulation(
        client_fn        = client_fn_fixed,
        num_clients      = int(NUM_CLIENTS),
        config           = fl.server.ServerConfig(num_rounds=ROUNDS),
        strategy         = strategy_obj,
        client_resources = {"num_cpus": 1, "num_gpus": 1},
        ray_init_args    = ray_init_args,
    )

    print("\n[run_fedsem] Simulation complete.")

    # 7) Save server history
    with open("fedsem_server_history.json", "w",
              encoding="utf-8") as f:
        json.dump(strategy_obj.server_history, f,
                  indent=2, default=json_default)
    print("[History] Saved fedsem_server_history.json")

    # 8) Save full FEDSEM_HIST
    save_fedsem_history("fedsem_full_history.json")
    print("[History] Saved fedsem_full_history.json")

    # 9) Post-simulation summary
    print_post_simulation_summary()

    # 10) Final evaluation on global clean test set
    final_ndarrays = FEDSEM_GLOBAL_FINAL.get("params", None)
    if final_ndarrays is None:
        print("[Warning] No aggregated params in "
              "FEDSEM_GLOBAL_FINAL. Skipping final eval.")
        return

    final_model = init_banglabert_classifier(num_labels=NUM_LABELS)
    set_model_weights(final_model, final_ndarrays)
    final_model.to(DEVICE)
    final_model.eval()

    global_test_loader = DataLoader(
        BanglaSentimentDataset(
            df_test_global,
            text_col  = "review_clean",
            label_col = "y",
            max_len   = ENC_MAX_LEN,
        ),
        batch_size = BATCH_SIZE,
        shuffle    = False,
    )

    m  = evaluate_all_metrics(
        final_model, global_test_loader, NUM_LABELS, DEVICE)
    cm = np.array(m["confusion_matrix"], dtype=int)

    print("\n" + "=" * 70)
    print("FINAL EVALUATION ON GLOBAL CLEAN TEST SET")
    print("=" * 70)
    print(f"Accuracy         : {_fmt(m.get('acc'))}")
    print(f"Macro Precision  : {_fmt(m.get('precision'))}")
    print(f"Macro Recall     : {_fmt(m.get('recall'))}")
    print(f"Macro F1         : {_fmt(m.get('f1'))}")
    print(f"Cohen's Kappa    : {_fmt(m.get('kappa'))}")
    print(f"ROC-AUC          : {_fmt(m.get('roc_auc'), nd=6)}")
    print("\nConfusion Matrix:")
    print(cm)

    final_metrics = {
        "accuracy":         float(m["acc"]),
        "precision":        float(m["precision"]),
        "recall":           float(m["recall"]),
        "f1":               float(m["f1"]),
        "kappa":            float(m["kappa"]),
        "roc_auc":          float(m["roc_auc"]),
        "confusion_matrix": m["confusion_matrix"],
    }

    with open("fedsem_final_metrics.json", "w",
              encoding="utf-8") as f:
        json.dump(final_metrics, f, indent=2)

    print("\n[History] Saved fedsem_final_metrics.json")
    print("\nSimulation and Evaluation completed successfully.")
```
